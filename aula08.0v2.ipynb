## 1. Por que criar funções?

Sem uma função, uma sequência usada várias vezes precisa ser repetida no código.

````python
print('Olá!')
print('Bem-vindo à aula de Python.')

print('Olá!')
print('Bem-vindo à aula de Python.')
````

Uma função permite dar um nome ao bloco e chamá-lo quantas vezes quisermos. Isso reduz repetição e ajuda a organizar o programa.

````python
def boas_vindas():
    print('Olá!')
    print('Bem-vindo à aula de Python.')

boas_vindas()
boas_vindas()
````

## 2. Definindo uma função

A estrutura básica é:

```python
def nome_da_funcao():
    instruções
```

- `def` informa que uma função será definida;
- o nome identifica a função;
- os parênteses recebem parâmetros, quando existirem;
- os dois-pontos iniciam o bloco;
- a indentação mostra quais instruções pertencem à função.

````python
def realizar_tarefa():
    print('Iniciando a tarefa...')
    print('Executando a tarefa...')
    print('Tarefa concluída.')
````

Executar a célula anterior apenas **define** a função. O bloco só é executado quando ela é chamada.

## 3. Chamando uma função

Para chamar uma função, escrevemos seu nome seguido de parênteses.

````python
realizar_tarefa()
````

Os parênteses fazem diferença:

- `realizar_tarefa` representa a própria função;
- `realizar_tarefa()` executa a função.

````python
outra_referencia = realizar_tarefa

print(f'É a mesma função? {outra_referencia is realizar_tarefa}')
outra_referencia()
````

A variável `outra_referencia` aponta para a mesma função. Essa possibilidade será útil ao estudarmos `map()` e `filter()`.

## 4. Parâmetros e argumentos

Parâmetros permitem que a mesma função trabalhe com valores diferentes.

- **parâmetro:** nome usado na definição da função;
- **argumento:** valor enviado no momento da chamada.

````python
def cumprimentar(nome):
    print(f'Olá, {nome}!')

cumprimentar('Ana')
cumprimentar('Bruno')
````

Na definição, `nome` é um parâmetro. Nas chamadas, `'Ana'` e `'Bruno'` são argumentos. Cada chamada reutiliza o mesmo bloco com um valor diferente.

### Mais de um parâmetro

Parâmetros são separados por vírgulas. Por padrão, os argumentos são associados pela posição.

````python
def apresentar_estudante(nome, turma):
    print(f'{nome} estuda na turma {turma}.')

apresentar_estudante('Carla', '2º ano A')
````

Trocar a ordem dos argumentos muda o significado. Podemos deixar a chamada explícita usando os nomes dos parâmetros.

````python
apresentar_estudante(turma='3º ano B', nome='Diego')
````

Esses são **argumentos nomeados**. Como o nome de cada parâmetro aparece, a ordem deixa de ser importante.

## 5. Retornando um resultado

A instrução `return` envia um valor de volta para o local onde a função foi chamada.

````python
def calcular_quadrado(numero):
    resultado = numero ** 2
    return resultado

quadrado_de_5 = calcular_quadrado(5)
print(f'Quadrado de 5: {quadrado_de_5}')
````

O valor retornado pode ser guardado em uma variável, usado em outro cálculo ou passado para outra função.

````python
resultado = calcular_quadrado(3) + calcular_quadrado(4)
print(f'3² + 4² = {resultado}')
````

### `return` encerra a função

Quando Python encontra `return`, devolve o valor e sai da função. Instruções colocadas depois dele no mesmo caminho não são executadas.

````python
def classificar_numero(numero):
    if numero > 0:
        return 'positivo'
    if numero < 0:
        return 'negativo'
    return 'zero'

print(classificar_numero(8))
print(classificar_numero(-2))
print(classificar_numero(0))
````

Cada chamada segue apenas um retorno. Por isso, após retornar `'positivo'`, a função não testa as linhas seguintes.

## 6. `print()` e `return` são diferentes

- `print()` mostra uma informação na tela;
- `return` entrega um valor para o restante do programa.

Mostrar um resultado não significa retorná-lo.

````python
def quadrado_com_print(numero):
    print(numero ** 2)

def quadrado_com_return(numero):
    return numero ** 2

retorno_do_print = quadrado_com_print(5)
retorno_real = quadrado_com_return(5)

print(f'Retorno da primeira função: {retorno_do_print}')
print(f'Retorno da segunda função: {retorno_real}')
````

A primeira função mostra `25`, mas não possui `return`; por isso, seu retorno é `None`, que representa ausência de valor. A segunda entrega `25`, que pode ser reutilizado.

## 8. Parâmetros com valor padrão

Um parâmetro pode ter um valor padrão. Ele será usado quando nenhum argumento correspondente for enviado.

````python
def cumprimentar_usuario(nome, saudacao='Olá'):
    return f'{saudacao}, {nome}!'

print(cumprimentar_usuario('Ana'))
print(cumprimentar_usuario('Bruno', 'Bom dia'))
````

Na primeira chamada, `saudacao` usa `'Olá'`. Na segunda, o argumento `'Bom dia'` substitui o padrão. Parâmetros sem valor padrão devem aparecer antes dos parâmetros com padrão.

## 9. Funções nativas do Python

Funções nativas, ou *built-ins*, já estão disponíveis sem que precisemos defini-las. Algumas foram usadas desde a primeira aula.

| Função | Recebe | Retorna ou produz |
|---|---|---|
| `print()` | valores para mostrar | mostra na tela; retorno `None` |
| `input()` | mensagem opcional | texto digitado (`str`) |
| `range()` | início, fim e passo | sequência de inteiros |
| `int()` | valor convertível | número inteiro |
| `float()` | valor convertível | número decimal |
| `str()` | um valor | representação em texto |
| `len()` | uma coleção | quantidade de itens |
| `type()` | um valor | seu tipo |

````python
texto_numerico = '42'
numero_inteiro = int(texto_numerico)
numero_decimal = float(texto_numerico)
texto_novamente = str(numero_inteiro)

print(numero_inteiro, type(numero_inteiro))
print(numero_decimal, type(numero_decimal))
print(texto_novamente, type(texto_novamente))
print(f'Tamanho do texto: {len(texto_novamente)}')
````

## 10. `sum()` e `zip()`

`sum()` soma os valores de uma sequência. `zip()` combina sequências pela posição, formando pares.

````python
notas = [8.0, 7.5, 9.0]
total = sum(notas)

print(f'Soma das notas: {total}')
````

````python
nomes = ['Ana', 'Bruno', 'Carla']
notas = [8.0, 7.5, 9.0]
pares = list(zip(nomes, notas))

print(pares)
for nome, nota in pares:
    print(f'{nome}: {nota}')
````

`zip()` combina até a menor sequência terminar. Usamos `list()` apenas para visualizar todos os pares de uma vez.

## 11. Funções como argumentos: `map()`

`map()` aplica uma função a cada item de uma sequência. Passamos a função sem parênteses, pois queremos entregá-la ao `map()`, não executá-la naquele momento.

````python
def calcular_cubo(numero):
    return numero ** 3

numeros = [1, 2, 3, 4]
cubos = list(map(calcular_cubo, numeros))

print(cubos)
````

`map()` devolve um objeto próprio para percorrer; `list()` transforma seus resultados em lista. A mesma tarefa também poderia ser feita com compreensão de listas.

## 12. Funções como argumentos: `filter()`

`filter()` mantém os itens para os quais uma função retorna `True`. Essa função funciona como um teste.

````python
def eh_par(numero):
    return numero % 2 == 0

numeros = [1, 2, 3, 4, 5, 6]
pares = list(filter(eh_par, numeros))

print(pares)
````

Assim como `map()`, `filter()` será retomado com funções `lambda` em uma aula posterior. Por enquanto, usamos funções nomeadas para deixar a lógica visível.

## 13. Exemplo completo: calculadora

Cada operação fica em uma função pequena. As funções recebem números e retornam resultados; a escolha do menu fica fora delas. Isso facilita testes e reutilização.

````python
def adicionar(a, b):
    return a + b

def subtrair(a, b):
    return a - b

def multiplicar(a, b):
    return a * b

def dividir(a, b):
    if b == 0:
        return None
    return a / b
````

````python
print(f'Adição: {adicionar(10, 2)}')
print(f'Subtração: {subtrair(10, 2)}')
print(f'Multiplicação: {multiplicar(10, 2)}')
print(f'Divisão: {dividir(10, 2)}')
print(f'Divisão por zero: {dividir(10, 0)}')
````

### Selecionando a operação

A próxima célula simula valores digitados. Para torná-la interativa, substitua as três primeiras atribuições pelas chamadas de `input()` mostradas nos comentários.

````python
# opcao = input('Operação (a/s/d/m): ').strip().lower()
# numero1 = float(input('Primeiro número: '))
# numero2 = float(input('Segundo número: '))
opcao = 'm'
numero1 = 6
numero2 = 7

if opcao == 'a':
    resultado = adicionar(numero1, numero2)
elif opcao == 's':
    resultado = subtrair(numero1, numero2)
elif opcao == 'm':
    resultado = multiplicar(numero1, numero2)
elif opcao == 'd':
    resultado = dividir(numero1, numero2)
else:
    resultado = 'Operação inválida'

if resultado is None:
    print('Não é possível dividir por zero.')
else:
    print(f'Resultado: {resultado}')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `def saudacao()` | faltaram os dois-pontos | `def saudacao():` |
| bloco sem indentação | Python não identifica o corpo | use quatro espaços |
| escrever apenas `saudacao` | a função não é chamada | use `saudacao()` |
| chamar antes de definir | o nome ainda não existe | execute primeiro a definição |
| enviar quantidade errada de argumentos | ocorre `TypeError` | confira os parâmetros |
| usar `print()` quando precisa reutilizar o resultado | o retorno será `None` | use `return` |
| usar `return` e esperar uma mensagem automática | retornar não mostra na tela | use `print(funcao(...))` |
| criar variável chamada `sum`, `str` ou `list` | encobre a função nativa | escolha outro nome |
| dividir sem verificar o divisor | zero provoca erro | trate `b == 0` |

### Uma solução para o desafio

````python
def calcular_media(notas):
    return sum(notas) / len(notas)

def classificar_media(media):
    if media >= 7:
        return 'Aprovado'
    return 'Reprovado'

def nota_aprovada(nota):
    return nota >= 7

def filtrar_notas_aprovadas(notas):
    return list(filter(nota_aprovada, notas))

def mostrar_relatorio(nome, notas):
    media = calcular_media(notas)
    classificacao = classificar_media(media)
    aprovadas = filtrar_notas_aprovadas(notas)

    print(f'--- {nome} ---')
    print(f'Notas: {notas}')
    print(f'Média: {media:.2f}')
    print(f'Classificação: {classificacao}')
    print(f'Notas maiores ou iguais a 7: {aprovadas}')

mostrar_relatorio('Ana', [8.0, 7.5, 9.0])
mostrar_relatorio('Bruno', [5.0, 6.0, 7.0])
````
