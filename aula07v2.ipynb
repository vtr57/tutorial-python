## 1. O que é um dicionário?

Um dicionário guarda associações entre **chaves** e **valores**. Pense em uma ficha de cadastro:

| Chave | Valor |
|---|---|
| `nome` | `Vitor` |
| `profissão` | `Professor` |

A chave funciona como uma etiqueta que permite encontrar seu valor.

````python
cadastro = {'nome': 'Vitor', 'profissão': 'Professor'}

print(cadastro)
print(type(cadastro))
````

Dicionários pertencem ao tipo `dict`. Eles são escritos entre chaves `{}`, com dois-pontos entre cada chave e valor e vírgulas entre os pares:

```python
dicionario = {chave1: valor1, chave2: valor2}
```

## 2. Lista ou dicionário?

| Estrutura | Como localizamos um valor? | Exemplo |
|---|---|---|
| lista | por posição ou índice | `aluno[0]` |
| dicionário | por chave | `aluno['nome']` |

Uma lista é adequada para uma sequência de itens parecidos. Um dicionário é adequado quando cada informação possui um significado identificado por uma chave.

````python
aluno_lista = ['Marina', 16, '2º ano B']
aluno_dicionario = {'nome': 'Marina', 'idade': 16, 'turma': '2º ano B'}

print(f'Na lista, aluno_lista[0]: {aluno_lista[0]}')
print(f"No dicionário, aluno_dicionario['nome']: {aluno_dicionario['nome']}")
````

A chave `'nome'` comunica melhor o significado do valor do que o índice `0`.

## 3. Criando dicionários

As chaves de um dicionário precisam ser únicas. Os valores podem ter tipos diferentes.

````python
livro = {
    'título': 'O Pequeno Príncipe',
    'ano': 1943,
    'preço': 29.90,
    'disponível': True
}

print(livro)
````

Quebrar um dicionário em várias linhas facilita a leitura. A vírgula depois do último par é opcional.

### Criando um dicionário vazio

Um dicionário vazio é criado com `{}` e pode ser preenchido depois.

````python
pessoa = {}

print(pessoa)
print(f'Quantidade de pares: {len(pessoa)}')
````

## 4. Acessando valores

Para acessar um valor, escrevemos o nome do dicionário e a chave entre colchetes.

````python
cadastro = {'nome': 'Vitor', 'profissão': 'Professor'}

print(f"Nome: {cadastro['nome']}")
print(f"Profissão: {cadastro['profissão']}")
````

### Chaves diferenciam maiúsculas e minúsculas

As chaves `'nome'`, `'Nome'` e `'NOME'` são diferentes. Use uma escrita consistente.

````python
exemplo = {'nome': 'Ana', 'Nome': 'Beatriz'}

print(exemplo['nome'])
print(exemplo['Nome'])
````

## 5. Adicionando pares

Para adicionar um par, usamos uma nova chave em uma atribuição:

```python
dicionario[nova_chave] = novo_valor
```

````python
cadastro = {'nome': 'Vitor', 'profissão': 'Professor'}
print(f'Antes: {cadastro}')

cadastro['salário'] = 1e4
cadastro['cidade'] = 'São Paulo'

print(f'Depois: {cadastro}')
print(f"Salário: {cadastro['salário']}")
````

O número `1e4` é outra forma de escrever `10000.0`, usando notação científica.

## 6. Atualizando valores

Se a chave já existe, a atribuição substitui seu valor em vez de criar outro par. Como as chaves são únicas, um dicionário não possui duas entradas separadas para a mesma chave.

````python
estoque = {'caderno': 12, 'caneta': 30}
print(f'Antes: {estoque}')

estoque['caderno'] = 20

print(f'Depois: {estoque}')
print(f"Quantidade de cadernos: {estoque['caderno']}")
````

Também podemos usar o valor atual em um cálculo.

````python
estoque = {'caderno': 12, 'caneta': 30}
estoque['caderno'] += 5
estoque['caneta'] -= 3

print(estoque)
````

## 7. Tamanho do dicionário

A função `len()` informa quantos pares chave-valor existem.

````python
aluno = {'nome': 'Marina', 'idade': 16, 'turma': '2º ano B'}

print(f'Quantidade de pares: {len(aluno)}')
````

## 8. Verificando se uma chave existe

O operador `in` verifica chaves. Isso permite consultar o dicionário apenas quando a chave estiver presente.

````python
aluno = {'nome': 'Marina', 'idade': 16}

print(f"'nome' está no dicionário? {'nome' in aluno}")
print(f"'turma' está no dicionário? {'turma' in aluno}")
````

````python
aluno = {'nome': 'Marina', 'idade': 16}
chave_procurada = 'turma'

if chave_procurada in aluno:
    print(aluno[chave_procurada])
else:
    print(f"A chave '{chave_procurada}' não foi encontrada.")
````

## 9. Consultas seguras com `.get()`

Acessar uma chave inexistente com colchetes provoca `KeyError`. O método `.get()` retorna o valor quando a chave existe e permite definir uma resposta padrão quando ela não existe.

```python
dicionario.get(chave, valor_padrao)
```

````python
aluno = {'nome': 'Marina', 'idade': 16}

nome = aluno.get('nome', 'Não informado')
turma = aluno.get('turma', 'Não informada')

print(f'Nome: {nome}')
print(f'Turma: {turma}')
````

Sem um valor padrão, `.get()` retorna `None` quando a chave não é encontrada. Para mensagens destinadas ao usuário, um texto padrão costuma ser mais claro.

## 10. Percorrendo dicionários com `for`

Podemos percorrer apenas as chaves, apenas os valores ou os pares completos.

### Percorrendo chaves com `.keys()`

````python
cadastro = {'nome': 'Vitor', 'profissão': 'Professor', 'cidade': 'São Paulo'}

for chave in cadastro.keys():
    print(chave)
````

Percorrer o próprio dicionário também percorre suas chaves.

````python
for chave in cadastro:
    print(f'{chave} -> {cadastro[chave]}')
````

### Percorrendo valores com `.values()`

````python
for valor in cadastro.values():
    print(valor)
````

### Percorrendo pares com `.items()`

`.items()` fornece cada par. O `for` desempacota o par em duas variáveis: uma recebe a chave e outra recebe o valor.

````python
for chave, valor in cadastro.items():
    print(f'{chave}: {valor}')
````

### Resumo dos métodos

| Método | O que percorre | Variáveis comuns |
|---|---|---|
| `.keys()` | chaves | `for chave in dicionario.keys()` |
| `.values()` | valores | `for valor in dicionario.values()` |
| `.items()` | pares | `for chave, valor in dicionario.items()` |

## 11. Desempacotando itens

Cada elemento produzido por `.items()` é um par com a forma `(chave, valor)`. Podemos distribuir alguns desses pares entre variáveis e reunir o restante com `*`.

````python
massas_atomicas = {'H': 1.008, 'He': 4.003, 'Li': 6.94, 'Be': 9.012, 'B': 10.81}
primeiro_par, segundo_par, *outros_pares = massas_atomicas.items()

print(f'Primeiro par: {primeiro_par}')
print(f'Segundo par: {segundo_par}')
print(f'Outros pares: {outros_pares}')
````

As variáveis recebem pares, e não apenas os valores. Para trabalhar com todos os elementos, o `for chave, valor in dicionario.items()` costuma ser mais simples e claro.

### Desempacotando uma cópia com `**`

Dois asteriscos podem inserir os pares de um dicionário em outro. Isso é útil para criar uma cópia com campos atualizados.

````python
cadastro_original = {'nome': 'Lia', 'cidade': 'Recife'}
cadastro_atualizado = {**cadastro_original, 'cidade': 'Olinda', 'idade': 17}

print(f'Original: {cadastro_original}')
print(f'Atualizado: {cadastro_atualizado}')
````

Como `'cidade'` aparece depois de `**cadastro_original`, o valor mais recente substitui o anterior apenas no novo dicionário.

## 12. Dicionários aninhados

Um valor também pode ser outro dicionário. Isso permite representar informações compostas, como contatos com telefone e e-mail.

````python
contatos = {
    'Ana': {'telefone': '1199999-1111', 'email': 'ana@exemplo.com'},
    'Bruno': {'telefone': '1199999-2222', 'email': 'bruno@exemplo.com'}
}

print(contatos)
````

Para chegar ao telefone de Ana, fazemos dois acessos: primeiro ao contato `'Ana'`, depois ao campo `'telefone'`.

````python
telefone_da_ana = contatos['Ana']['telefone']
email_do_bruno = contatos['Bruno']['email']

print(f'Telefone de Ana: {telefone_da_ana}')
print(f'E-mail de Bruno: {email_do_bruno}')
````

## 13. Exemplo completo: estoque de uma loja

Neste exemplo, as chaves são produtos e os valores são preços. Depois de adicionar e atualizar itens, percorremos o catálogo.

````python
precos = {
    'caderno': 18.50,
    'caneta': 3.00,
    'mochila': 120.00
}

precos['régua'] = 4.50
precos['caneta'] = 3.25

print(f'Quantidade de produtos: {len(precos)}')
for produto, preco in precos.items():
    print(f'{produto}: R$ {preco:.2f}')
````

Em `{preco:.2f}`, a parte `.2f` formata o número com duas casas decimais.

## 14. Exemplo completo: consulta em um glossário

## 15. Exemplo completo: calculando um pedido

O menu associa pratos a preços. A lista `pedido` simula as escolhas de uma pessoa. O laço consulta cada item e acumula o total.

````python
menu = {
    'sanduíche': 18.00,
    'suco': 7.50,
    'salada': 16.00
}
pedido = ['sanduíche', 'suco']
total = 0

for item in pedido:
    if item in menu:
        preco = menu[item]
        total += preco
        print(f'{item}: R$ {preco:.2f}')
    else:
        print(f'{item}: indisponível')

print(f'Total: R$ {total:.2f}')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `{'nome' = 'Ana'}` | pares usam dois-pontos | `{'nome': 'Ana'}` |
| `cadastro[nome]` | sem aspas, Python procura uma variável | `cadastro['nome']` |
| acessar uma chave inexistente com `[]` | ocorre `KeyError` | teste com `in` ou use `.get()` |
| esperar duas chaves `'nome'` | chaves devem ser únicas | use chaves diferentes ou atualize o valor |
| `for chave, valor in cadastro` | o dicionário sozinho fornece chaves | use `cadastro.items()` |
| confundir `.keys()` com `.values()` | um fornece etiquetas e outro dados | escolha o método conforme o objetivo |
| usar índice numérico sem que ele seja uma chave | dicionários não usam posições como listas | acesse pela chave real |
