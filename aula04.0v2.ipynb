## 1. Por que usar uma lista?

Imagine que queremos guardar as temperaturas medidas em três horários. Poderíamos criar uma variável para cada valor:

````python
temperatura_1 = 20.5
temperatura_2 = 23.0
temperatura_3 = 21.8

print(temperatura_1, temperatura_2, temperatura_3)
````

Isso funciona para poucos valores, mas ficaria cansativo com dezenas de medições. Uma lista permite guardar todos os valores juntos e manter sua ordem.

````python
temperaturas = [20.5, 23.0, 21.8]
print(temperaturas)
print(type(temperaturas))
````

## 2. Criando uma lista

Uma lista é escrita entre colchetes `[]`. Seus itens são separados por vírgulas.

```python
nome_da_lista = [item1, item2, item3]
```

A lista mantém os itens na ordem em que foram colocados.

````python
notas = [8.0, 7.5, 9.2]
frutas = ['banana', 'maçã', 'laranja']

print(notas)
print(frutas)
````

### Lista e item não são a mesma coisa

Na lista `['banana', 'maçã', 'laranja']`:

- a lista inteira é um valor do tipo `list`;
- cada fruta é um item do tipo `str`.

````python
frutas = ['banana', 'maçã', 'laranja']

print(type(frutas))
print(type(frutas[0]))
````

### Tipos dos itens

Python permite misturar tipos dentro de uma lista, como `['Ana', 16, 8.5]`. Isso pode ser útil, mas listas com itens da mesma categoria costumam ser mais fáceis de entender e usar.

````python
dados_estudante = ['Ana', 16, 8.5]
print(dados_estudante)
````

## 3. Descobrindo o tamanho com `len()`

A função `len()` informa quantos itens existem em uma lista. A palavra vem de *length*, que significa comprimento.

````python
materias = ['Matemática', 'Português', 'História', 'Biologia']
quantidade_de_materias = len(materias)

print(materias)
print(f'A lista possui {quantidade_de_materias} matérias.')
````

## 4. Acessando itens por índice

Cada posição de uma lista possui um número chamado **índice**. Em Python, a contagem começa em zero.

| Item | `'segunda'` | `'terça'` | `'quarta'` | `'quinta'` | `'sexta'` |
|---|---:|---:|---:|---:|---:|
| Índice | `0` | `1` | `2` | `3` | `4` |

Para acessar um item, escrevemos o nome da lista seguido do índice entre colchetes: `lista[indice]`.

````python
dias_uteis = ['segunda', 'terça', 'quarta', 'quinta', 'sexta']

print(f'Primeiro item: {dias_uteis[0]}')
print(f'Segundo item: {dias_uteis[1]}')
print(f'Terceiro item: {dias_uteis[2]}')
````

### Posição humana e índice

No cotidiano, chamamos um item de “primeiro”. Em Python, o índice desse item é `0`.

| Posição | Índice |
|---:|---:|
| primeiro item | `0` |
| segundo item | `1` |
| terceiro item | `2` |

> Para acessar a posição desejada, normalmente subtraímos 1: quinto item → índice `4`.

### Acessando o último item

O índice `-1` representa o último item da lista. O índice `-2` representa o penúltimo, e assim por diante.

````python
planetas_rochosos = ['Mercúrio', 'Vênus', 'Terra', 'Marte']

print(f'Primeiro planeta: {planetas_rochosos[0]}')
print(f'Último planeta: {planetas_rochosos[-1]}')
print(f'Penúltimo planeta: {planetas_rochosos[-2]}')
````

## 5. Erro de índice: `IndexError`

Considere uma lista com três itens:

```python
dia = [30.2, 55, 20]
```

Os índices válidos são `0`, `1` e `2`. Tentar acessar `dia[3]` gera `IndexError: list index out of range`, porque o índice `3` ainda não existe.

> A célula com erro não será executada aqui, para que o notebook possa rodar do início ao fim.

````python
dia = [30.2, 55, 20]

print(f'Tamanho da lista: {len(dia)}')
print('Índices válidos: 0, 1 e 2')
print(f'Último item usando -1: {dia[-1]}')
````

### Como evitar o erro

Antes de usar um índice, observe o tamanho da lista com `len()`. Em uma lista de tamanho `4`, os índices positivos vão de `0` até `3`. O maior índice válido é sempre `len(lista) - 1`.

````python
cores = ['azul', 'verde', 'amarelo', 'roxo']
ultimo_indice = len(cores) - 1

print(f'Tamanho: {len(cores)}')
print(f'Último índice: {ultimo_indice}')
print(f'Último item: {cores[ultimo_indice]}')
````

## 6. Adicionando itens com `append()`

O método `append()` adiciona um item ao final da lista.

```python
lista.append(novo_item)
```

A lista é modificada e passa a ter um item a mais.

````python
dia = [30.2, 55, 20]
print(f'Antes: {dia}')
print(f'Tamanho antes: {len(dia)}')

dia.append(72)

print(f'Depois: {dia}')
print(f'Tamanho depois: {len(dia)}')
print(f'Novo último item: {dia[-1]}')
````

Podemos chamar `append()` várias vezes. Cada chamada adiciona exatamente um item ao final.

````python
tarefas = ['estudar Matemática']
tarefas.append('ler um capítulo')
tarefas.append('organizar o material')

print(tarefas)
````

### `append()` altera a lista

Não precisamos escrever `tarefas = tarefas.append(...)`. Basta chamar `tarefas.append(...)`, pois o método modifica a própria lista.

## 7. Alterando um item

Listas podem ser modificadas depois de criadas. Para trocar um item, usamos seu índice e uma atribuição:

```python
lista[indice] = novo_valor
```

````python
cores = ['azul', 'verde', 'amarelo']
print(f'Antes: {cores}')

cores[1] = 'vermelho'

print(f'Depois: {cores}')
print(f'Nova segunda cor: {cores[1]}')
````

### Alterar não é adicionar

- `cores[1] = 'vermelho'` substitui o item que já ocupa o índice `1`;
- `cores.append('vermelho')` cria um novo item no final.

Ao alterar um item, o tamanho da lista permanece igual. Ao usar `append()`, o tamanho aumenta em um.

````python
numeros = [10, 20, 30]

numeros[0] = 15
print(f'Após alterar: {numeros} | tamanho: {len(numeros)}')

numeros.append(40)
print(f'Após adicionar: {numeros} | tamanho: {len(numeros)}')
````

## 8. Criando uma lista vazia

Uma lista vazia é criada com `[]`. Ela começa sem itens e pode ser preenchida aos poucos com `append()`.

````python
dias_da_semana = []

print(dias_da_semana)
print(f'Tamanho inicial: {len(dias_da_semana)}')
````

````python
dias_da_semana = []
dias_da_semana.append('segunda-feira')
dias_da_semana.append('terça-feira')
dias_da_semana.append('quarta-feira')

print(dias_da_semana)
print(f'Tamanho atual: {len(dias_da_semana)}')
````

### Cuidado com listas vazias

Uma lista vazia tem tamanho zero e ainda não possui nem mesmo o índice `0`. Portanto, não tente acessar ou alterar `lista[0]` antes de adicionar um item.

## 9. Exemplo completo: plano de estudos

Vamos criar uma lista, consultar seus itens, adicionar uma matéria e corrigir outra.

````python
plano_de_estudos = ['Matemática', 'Geografia', 'Fisica']

print(f'Plano inicial: {plano_de_estudos}')
print(f'Primeira matéria: {plano_de_estudos[0]}')
print(f'Última matéria: {plano_de_estudos[-1]}')

plano_de_estudos.append('Programação')
plano_de_estudos[2] = 'Física'

print(f'Plano atualizado: {plano_de_estudos}')
print(f'Quantidade de matérias: {len(plano_de_estudos)}')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `frutas = banana, maçã` | faltaram colchetes e aspas | `frutas = ['banana', 'maçã']` |
| `frutas[1]` para acessar o primeiro item | a contagem começa em zero | use `frutas[0]` |
| `frutas[5]` em uma lista de três itens | o índice não existe | confira o tamanho com `len()` |
| `frutas.append = 'uva'` | faltou chamar o método | use `frutas.append('uva')` |
| `frutas = frutas.append('uva')` | `append()` já modifica a lista | use apenas `frutas.append('uva')` |
| `lista[0] = 'item'` em uma lista vazia | ainda não existe índice `0` | use `lista.append('item')` |

Ao encontrar `IndexError`, confira o tamanho da lista e lembre que os índices começam em zero.
