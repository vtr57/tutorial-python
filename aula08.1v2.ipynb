## 1. O que é escopo?

Escopo é a região do programa em que um nome pode ser encontrado. Nesta aula, trabalharemos principalmente com dois níveis:

- **escopo global:** nomes criados fora das funções;
- **escopo local:** nomes criados dentro de uma função.

Funções criam seu próprio espaço local. Isso permite que diferentes partes do programa usem nomes iguais sem alterar umas às outras.

## 2. Variável global

Uma variável criada fora de qualquer função está no escopo global do notebook. Ela pode ser usada pelas células executadas depois de sua criação.

````python
escola = 'Escola Python'
ano = 2026

print(f'Escola: {escola}')
print(f'Ano: {ano}')
````

As variáveis `escola` e `ano` foram criadas fora de funções, por isso são globais.

## 3. Variável local

Uma variável criada dentro de uma função é local. Ela pode ser usada durante aquela chamada, dentro do bloco da função.

````python
def calcular_area():
    base = 8
    altura = 5
    area = base * altura
    print(f'Dentro da função, área = {area}')

calcular_area()
````

`base`, `altura` e `area` são locais. Fora da função, esses nomes não foram criados. Tentar executar `print(area)` fora dela provocaria `NameError`.

> O código com erro não será executado, para que o notebook continue funcionando do início ao fim.

### Como levar um valor para fora

Se o restante do programa precisa do resultado, a função deve retorná-lo.

````python
def calcular_area():
    base = 8
    altura = 5
    area = base * altura
    return area

area_calculada = calcular_area()
print(f'Fora da função, retorno = {area_calculada}')
````

A variável local `area` continua restrita à função. `area_calculada` é uma nova variável global que recebe o valor retornado.

## 4. Parâmetros são locais

O parâmetro pertence ao escopo local da função. O argumento fornece seu valor em cada chamada.

````python
def calcular_quadrado(numero):
    resultado = numero ** 2
    return resultado

print(calcular_quadrado(3))
print(calcular_quadrado(5))
````

Em cada chamada, o parâmetro local `numero` recebe um argumento diferente. O nome não continua disponível fora da função.

## 5. Mesmo nome em escopos diferentes

Uma variável local pode ter o mesmo nome de uma variável global. Dentro da função, o nome local tem prioridade e **sombreia** temporariamente o global. Os dois valores são independentes.

````python
x = 'Fora da função, x é um texto.'

def mostrar_x(x=5):
    print(f'Dentro da função, x = {x}')

mostrar_x(x=10)
print(f'Fora da função, x = {x}')
````

Neste exemplo:

- `x` fora da função é uma variável global com texto;
- `x` na definição é um parâmetro local com padrão `5`;
- `10` é o argumento da chamada;
- o parâmetro recebe `10`, mas a variável global continua com o texto.

````python
valor = 100

def alterar_localmente():
    valor = 25
    print(f'Valor local: {valor}')

alterar_localmente()
print(f'Valor global: {valor}')
````

A atribuição `valor = 25` cria uma variável local. Ela não atualiza a variável global `valor = 100`.

## 6. Funções possuem locais independentes

Duas funções podem usar o mesmo nome local sem conflito.

````python
def primeira_funcao():
    mensagem = 'Mensagem da primeira função'
    print(mensagem)

def segunda_funcao():
    mensagem = 'Mensagem da segunda função'
    print(mensagem)

primeira_funcao()
segunda_funcao()
````

Cada `mensagem` existe apenas durante a execução de sua própria função.

## 7. Lendo uma variável global

Uma função pode ler uma variável global quando não existe um nome local igual.

````python
y = 10

def mostrar_valores():
    z = 20
    return f'z é local e vale {z}; y é global e vale {y}.'

mensagem = mostrar_valores()
print(mensagem)
````

A função encontra `z` localmente e encontra `y` no escopo global. Depois da chamada, `z` deixa de estar disponível; executar `print(z)` fora da função provocaria `NameError`.

## 8. Ordem simplificada de busca

Quando encontra um nome dentro de uma função, Python procura nesta ordem simplificada:

1. escopo local da função;
2. escopo global do módulo ou notebook;
3. nomes nativos, como `len`, `sum` e `print`.

Se o nome não for encontrado, ocorre `NameError`. Funções dentro de outras funções acrescentam um nível intermediário, que não será necessário nesta aula.

````python
categoria = 'global'

def descobrir_categoria():
    categoria = 'local'
    return categoria

print(f'Dentro da função: {descobrir_categoria()}')
print(f'Fora da função: {categoria}')
print(f'Nome nativo encontrado: {len([1, 2, 3])}')
````

## 9. Ler não é o mesmo que alterar

Ler uma global dentro da função é permitido. Porém, uma atribuição normalmente cria um nome local. Por isso, o código abaixo provocaria `UnboundLocalError`:

```python
contador = 0

def incrementar():
    contador += 1
```

Como há uma atribuição, Python trata `contador` como local, mas tenta ler seu valor local antes que ele exista. O código não será executado aqui.

## 10. Alterando uma global com `global`

A instrução `global` informa que uma atribuição dentro da função deve atingir o nome global.

````python
contador_global = 0

def incrementar_global():
    global contador_global
    contador_global += 1

print(f'Antes: {contador_global}')
incrementar_global()
incrementar_global()
print(f'Depois: {contador_global}')
````

`global` funciona, mas cria uma dependência: o resultado da função passa a depender de um estado externo e também o modifica. Isso pode dificultar testes e a compreensão do programa.

## 11. Alternativa preferível: parâmetro e retorno

Em muitos casos, podemos receber o valor por parâmetro e devolver o novo valor. A função fica independente de uma variável global específica.

````python
def incrementar(contador):
    return contador + 1

contador = 0
print(f'Antes: {contador}')

contador = incrementar(contador)
contador = incrementar(contador)

print(f'Depois: {contador}')
````

A função apenas calcula e retorna. O código que a chamou decide onde guardar o resultado. Essa separação geralmente torna o comportamento mais previsível.

````python
def aplicar_bonus(pontuacao, bonus):
    nova_pontuacao = pontuacao + bonus
    return nova_pontuacao

pontuacao = 100
pontuacao_atualizada = aplicar_bonus(pontuacao, 25)

print(f'Pontuação original: {pontuacao}')
print(f'Pontuação retornada: {pontuacao_atualizada}')
````

A variável global `pontuacao` não foi alterada automaticamente. O retorno foi guardado em outra variável, deixando a mudança explícita.

## 12. Tempo de vida de uma variável local

As variáveis locais são criadas em cada chamada. Chamadas diferentes podem produzir valores locais diferentes sem interferência.

````python
def criar_mensagem(nome):
    mensagem_local = f'Olá, {nome}!'
    return mensagem_local

mensagem_ana = criar_mensagem('Ana')
mensagem_bruno = criar_mensagem('Bruno')

print(mensagem_ana)
print(mensagem_bruno)
````

Cada chamada cria seu próprio parâmetro `nome` e sua própria `mensagem_local`. Apenas os textos retornados são guardados globalmente em `mensagem_ana` e `mensagem_bruno`.

## 13. Exemplo completo: previsão de valores

Leia o código e tente prever as três linhas antes de executar.

````python
numero = 10

def transformar(numero=20):
    print(f'Parâmetro ao entrar: {numero}')
    numero = 40
    print(f'Variável local após atribuição: {numero}')
    return numero

retorno = transformar(30)
print(f'Retorno: {retorno}')
print(f'Variável global: {numero}')
````

A chamada envia `30` ao parâmetro local. Em seguida, a atribuição local troca esse valor por `40`, que é retornado. A variável global `numero` continua valendo `10`.

## 14. Exemplo completo: configuração global e cálculo local

Constantes de configuração podem ser lidas globalmente. Por convenção, seus nomes são escritos em maiúsculas para indicar que não devem ser alterados durante o programa.

````python
TAXA_DE_DESCONTO = 0.10

def calcular_preco_com_desconto(preco):
    desconto_local = preco * TAXA_DE_DESCONTO
    preco_final = preco - desconto_local
    return preco_final

print(f'Preço final: R$ {calcular_preco_com_desconto(100):.2f}')
print(f'Taxa global: {TAXA_DE_DESCONTO}')
````

A função lê `TAXA_DE_DESCONTO`, mas não a altera. `desconto_local` e `preco_final` pertencem somente à função.

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| usar fora da função uma variável criada dentro | o nome local não existe fora | retorne o valor |
| achar que um parâmetro altera a global de mesmo nome | o parâmetro é local | guarde explicitamente o retorno |
| fazer `contador += 1` sem parâmetro nem `global` | ocorre `UnboundLocalError` | receba e retorne ou declare `global` |
| usar `global` para todos os resultados | funções ficam dependentes do estado externo | prefira parâmetros e retornos |
| criar variável chamada `len` ou `sum` | o nome local ou global esconde a função nativa | escolha outro nome |
| confiar em uma global sem deixar isso claro | o resultado pode mudar com estado externo | passe o valor como argumento quando possível |

### Uma solução para o desafio
