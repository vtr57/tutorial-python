## 1. Por que usar objetos?

Imagine um carro. Ele possui informações e ações relacionadas:

- **dados:** cor, ano, marca, modelo e velocidade;
- **comportamentos:** ligar, desligar, acelerar e frear.

A POO permite reunir esses dados e comportamentos em um objeto, em vez de espalhá-los em várias variáveis e funções sem uma ligação explícita.

### Vocabulário essencial

| Termo | Significado | Exemplo |
|---|---|---|
| classe | molde que descreve objetos | `Carro` |
| objeto ou instância | elemento criado a partir da classe | `carro_1` |
| atributo | dado pertencente ao objeto | `carro_1.cor` |
| método | função pertencente à classe | `carro_1.acelerar(20)` |
| estado | conjunto atual dos atributos | velocidade igual a 20 |

## 2. Classe e objeto

Uma classe funciona como uma descrição ou molde. Os objetos são elementos concretos criados a partir dela.

Uma planta de casa não é uma casa habitável; ela descreve como casas podem ser construídas. Da mesma forma, uma classe descreve objetos, mas precisamos instanciá-la para criar cada objeto.

## 3. Criando a primeira classe

Usamos a palavra `class` e, por convenção, nomes de classes começam com letra maiúscula.

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade
````

Executar a célula define a classe, mas ainda não cria nenhuma pessoa.

## 4. O método `__init__`

`__init__` é chamado automaticamente quando um objeto é criado. Ele configura o estado inicial do objeto.

```python
def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade
```

Os parâmetros `nome` e `idade` recebem os argumentos da criação. As atribuições guardam esses valores como atributos do novo objeto.

## 5. O papel de `self`

`self` representa o objeto que está sendo construído ou usado naquele momento.

- `nome` é um parâmetro local do método;
- `self.nome` é um atributo que permanece no objeto.

O nome `self` é uma convenção muito importante em Python. Ele deve ser o primeiro parâmetro dos métodos de instância.

## 6. Instanciando um objeto

Instanciar significa criar um objeto a partir de uma classe. Chamamos a classe como se fosse uma função.

````python
pessoa_1 = Pessoa(nome='Ana', idade=16)

print(pessoa_1)
print(type(pessoa_1))
````

A saída direta mostra que `pessoa_1` é um objeto da classe `Pessoa`. Para consultar seus dados, acessamos os atributos.

## 7. Acessando atributos

Usamos um ponto entre o objeto e o nome do atributo.

````python
print(f'Nome: {pessoa_1.nome}')
print(f'Idade: {pessoa_1.idade}')
````

A expressão `pessoa_1.nome` pode ser lida como “o atributo nome do objeto pessoa_1”.

## 8. Vários objetos da mesma classe

A mesma classe pode criar muitos objetos. Cada um possui seus próprios atributos.

````python
pessoa_2 = Pessoa('Bruno', 17)
pessoa_3 = Pessoa('Carla', 16)

print(f'{pessoa_1.nome}: {pessoa_1.idade} anos')
print(f'{pessoa_2.nome}: {pessoa_2.idade} anos')
print(f'{pessoa_3.nome}: {pessoa_3.idade} anos')
````

Alterar um objeto não muda automaticamente os demais.

````python
pessoa_2.idade = 18

print(f'Idade de Ana: {pessoa_1.idade}')
print(f'Idade de Bruno: {pessoa_2.idade}')
````

## 9. Criando métodos

Métodos são funções definidas dentro da classe. Eles normalmente usam `self` para consultar ou alterar o objeto. Vamos acrescentar uma saudação à classe `Pessoa`.

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def cumprimentar(self):
        return f'Olá! Meu nome é {self.nome}.'

pessoa = Pessoa('Davi', 17)
print(pessoa.cumprimentar())
````

Chamamos o método com `pessoa.cumprimentar()`. Não enviamos `self` manualmente; Python associa o objeto `pessoa` a esse parâmetro.

### Método com parâmetros adicionais

Além de `self`, um método pode receber outros parâmetros.

````python
class Contador:
    def __init__(self, valor_inicial=0):
        self.valor = valor_inicial

    def incrementar(self, quantidade):
        self.valor += quantidade
        return self.valor

contador = Contador(10)
print(contador.incrementar(5))
print(contador.incrementar(2))
````

O atributo `valor` guarda o estado entre chamadas. Depois de somar 5, ele vale 15; a chamada seguinte começa desse estado e soma 2.

## 10. Exemplo principal: classe `Carro`

A classe reúne características e comportamentos de um carro. `velocidade` começa em zero para todos os novos objetos.

````python
class Carro:
    def __init__(self, cor, ano, marca, modelo):
        self.cor = cor
        self.ano = ano
        self.marca = marca
        self.modelo = modelo
        self.velocidade = 0

    def ligar(self):
        return f'{self.marca} {self.modelo}: carro ligado'

    def desligar(self):
        return f'{self.marca} {self.modelo}: carro desligado'

    def acelerar(self, aceleracao):
        if aceleracao > 0:
            self.velocidade += aceleracao
        return self.velocidade

    def frear(self, desaceleracao):
        if desaceleracao > 0:
            self.velocidade -= desaceleracao
            if self.velocidade < 0:
                self.velocidade = 0
        return self.velocidade

    def exibir_status(self):
        return f'{self.marca} {self.modelo} - {self.velocidade} km/h'
````

### Instanciando um carro

````python
carro_1 = Carro(ano=2020, cor='preto', marca='Audi', modelo='A8')

print(carro_1.exibir_status())
````

Os argumentos nomeados deixam explícito qual valor corresponde a cada parâmetro. A ordem pode ser diferente da definição.

### Alterando o estado com métodos

````python
print(carro_1.ligar())
print(f'Após acelerar: {carro_1.acelerar(70)} km/h')
print(f'Após frear: {carro_1.frear(20)} km/h')
print(carro_1.exibir_status())
print(carro_1.desligar())
````

`acelerar()` e `frear()` alteram `self.velocidade`. O atributo guarda o novo estado para as próximas chamadas.

### Validando o estado

A frenagem impede uma velocidade negativa.

````python
print(f'Antes da frenagem forte: {carro_1.velocidade} km/h')
carro_1.frear(100)
print(f'Depois da frenagem forte: {carro_1.velocidade} km/h')
````

A classe protege uma regra do próprio objeto: a velocidade mínima é zero.

## 11. Objetos independentes

Cada carro criado possui seu próprio estado.

````python
carro_a = Carro('vermelho', 2022, 'Toyota', 'Corolla')
carro_b = Carro('azul', 2024, 'Honda', 'Civic')

carro_a.acelerar(40)
carro_b.acelerar(80)
carro_b.frear(10)

print(carro_a.exibir_status())
print(carro_b.exibir_status())
````

Os dois objetos compartilham os mesmos métodos definidos pela classe, mas seus atributos possuem valores independentes.

## 12. Percorrendo uma lista de objetos

Objetos podem ser guardados em listas e percorridos com `for`.

````python
carros = [carro_1, carro_a, carro_b]

for carro in carros:
    print(carro.exibir_status())
````

A variável `carro` recebe um objeto por vez. Em cada iteração, chamamos o mesmo método no objeto atual.

## 13. Verificando a classe de um objeto

A função `isinstance(objeto, Classe)` verifica se um objeto foi criado a partir de determinada classe.

````python
print(isinstance(carro_a, Carro))
print(isinstance(pessoa, Carro))
print(isinstance(pessoa, Pessoa))
````

O resultado é booleano. Herança ampliará esse comportamento na Aula 09.1.

## 14. Exemplo: classe `Circulo`

Uma classe também pode representar um conceito matemático. A área e o perímetro usam o atributo `raio`.

````python
from math import pi

class Circulo:
    def __init__(self, raio):
        self.raio = raio

    def calcular_area(self):
        return pi * self.raio ** 2

    def calcular_perimetro(self):
        return 2 * pi * self.raio

circulo = Circulo(3)
print(f'Área: {circulo.calcular_area():.2f}')
print(f'Perímetro: {circulo.calcular_perimetro():.2f}')
````

O raio é um dado do círculo; área e perímetro são comportamentos calculados a partir desse estado.

## 15. Exemplo: conta bancária

Métodos podem validar regras antes de alterar o estado.

````python
class ContaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        if valor <= 0:
            return False
        self.saldo += valor
        return True

    def sacar(self, valor):
        if valor <= 0 or valor > self.saldo:
            return False
        self.saldo -= valor
        return True

    def consultar_saldo(self):
        return self.saldo
````

````python
conta = ContaBancaria('Ana', 100)

print(f'Depósito realizado? {conta.depositar(50)}')
print(f'Saque realizado? {conta.sacar(30)}')
print(f'Saque acima do saldo? {conta.sacar(500)}')
print(f'Saldo final: R$ {conta.consultar_saldo():.2f}')
````

Os métodos retornam `True` ou `False` para informar se a operação foi aceita. O saldo só muda em operações válidas.

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `class pessoa:` | nome de classe não segue a convenção | use `class Pessoa:` |
| escrever `_init_` | o nome especial possui dois `_` de cada lado | use `__init__` |
| esquecer `self` no método | o objeto não é recebido | coloque `self` primeiro |
| `nome = nome` no `__init__` | o valor fica apenas em variável local | use `self.nome = nome` |
| chamar `Pessoa()` sem os argumentos exigidos | ocorre `TypeError` | envie os valores necessários |
| escrever `objeto.metodo` esperando execução | faltaram parênteses | use `objeto.metodo()` |
| chamar `objeto.metodo(objeto)` | `self` é enviado automaticamente | envie apenas os demais argumentos |
| usar `velocidade` em vez de `self.velocidade` | o estado do objeto não é alterado | acesse pelo `self` |

## 16. Prática guiada

Altere os dados, crie outro objeto e execute uma sequência diferente de métodos.

````python
class Jogo:
    def __init__(self, titulo, pontuacao=0):
        self.titulo = titulo
        self.pontuacao = pontuacao

    def marcar_pontos(self, pontos):
        if pontos > 0:
            self.pontuacao += pontos
        return self.pontuacao

    def exibir_status(self):
        return f'{self.titulo}: {self.pontuacao} pontos'

jogo = Jogo('Desafio Python', 10)
jogo.marcar_pontos(20)
jogo.marcar_pontos(15)
print(jogo.exibir_status())
````

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def cumprimentar(self):
        return f'Olá! Meu nome é {self.nome}.'

class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def descrever(self):
        return f'{self.nome}: R$ {self.preco:.2f}'

ana = Pessoa('Ana', 16)
bruno = Pessoa('Bruno', 17)
print(ana.nome, ana.idade, ana.cumprimentar())
print(bruno.nome, bruno.idade, bruno.cumprimentar())

produto_1 = Produto('Caderno', 18.5)
produto_2 = Produto('Caneta', 3.25)
produto_1.preco = 20.0
print(produto_1.descrever())
print(produto_2.descrever())
````

### Gabarito do Nível 2

````python
from math import pi

class Circulo:
    def __init__(self, raio):
        self.raio = raio

    def calcular_area(self):
        return pi * self.raio ** 2

    def calcular_perimetro(self):
        return 2 * pi * self.raio

class Retangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def calcular_area(self):
        return self.base * self.altura

    def calcular_perimetro(self):
        return 2 * (self.base + self.altura)

class Contador:
    def __init__(self, valor=0):
        self.valor = valor

    def incrementar(self, quantidade=1):
        if quantidade > 0:
            self.valor += quantidade
        return self.valor

    def reduzir(self, quantidade=1):
        if quantidade > 0:
            self.valor -= quantidade
            if self.valor < 0:
                self.valor = 0
        return self.valor

circulo = Circulo(2)
retangulo = Retangulo(4, 3)
contador = Contador(2)
contador.reduzir(10)

assert round(circulo.calcular_area(), 2) == round(4 * pi, 2)
assert retangulo.calcular_area() == 12
assert contador.valor == 0
print('Testes do Nível 2 aprovados.')
````

### Gabarito do Nível 3

````python
class Livro:
    def __init__(self, titulo, autor, ano):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano

    def obter_informacoes(self):
        return f'{self.titulo}, de {self.autor} ({self.ano})'

class ContaBancaria:
    def __init__(self, titular, saldo=0):
        self.titular = titular
        self.saldo = saldo

    def depositar(self, valor):
        if valor <= 0:
            return False
        self.saldo += valor
        return True

    def sacar(self, valor):
        if valor <= 0 or valor > self.saldo:
            return False
        self.saldo -= valor
        return True

livros = [
    Livro('O Pequeno Príncipe', 'Antoine de Saint-Exupéry', 1943),
    Livro('Quarto de Despejo', 'Carolina Maria de Jesus', 1960),
    Livro('Capitães da Areia', 'Jorge Amado', 1937)
]
for livro in livros:
    print(livro.obter_informacoes())

conta_1 = ContaBancaria('Ana', 100)
conta_2 = ContaBancaria('Bruno', 50)
conta_1.depositar(40)
conta_2.sacar(20)
assert conta_1.saldo == 140
assert conta_2.saldo == 30
print(f'{conta_1.titular}: R$ {conta_1.saldo:.2f}')
print(f'{conta_2.titular}: R$ {conta_2.saldo:.2f}')
````

### Uma solução para o desafio

````python
class Estudante:
    def __init__(self, nome):
        self.nome = nome
        self.notas = []

    def adicionar_nota(self, nota):
        if nota < 0 or nota > 10:
            return False
        self.notas.append(nota)
        return True

    def calcular_media(self):
        if len(self.notas) == 0:
            return 0
        return sum(self.notas) / len(self.notas)

    def obter_situacao(self):
        if self.calcular_media() >= 7:
            return 'Aprovado'
        return 'Reprovado'

    def exibir_relatorio(self):
        media = self.calcular_media()
        return (f'{self.nome} | notas: {self.notas} | '
                f'média: {media:.2f} | situação: {self.obter_situacao()}')

ana = Estudante('Ana')
bruno = Estudante('Bruno')
carla = Estudante('Carla')

for nota in [8.0, 7.5, 9.0]:
    ana.adicionar_nota(nota)
for nota in [5.0, 6.0, 7.0]:
    bruno.adicionar_nota(nota)
for nota in [7.0, 7.0, 7.0]:
    carla.adicionar_nota(nota)

estudantes = [ana, bruno, carla]
for estudante in estudantes:
    print(estudante.exibir_relatorio())
````
