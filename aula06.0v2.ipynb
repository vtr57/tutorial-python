## 1. Por que usar estruturas de repetição?

Para mostrar os números de 1 a 5, poderíamos escrever cinco chamadas de `print()`:

````python
print(1)
print(2)
print(3)
print(4)
print(5)
````

Isso funciona, mas repete código e fica pouco prático para centenas de valores. Uma estrutura de repetição, também chamada de **laço** ou *loop*, executa um bloco várias vezes. Cada execução é chamada de **iteração**.

## 2. Repetição com `while`

`while` significa “enquanto”. O bloco é repetido enquanto a condição for verdadeira.

```python
while condição:
    instruções repetidas
```

Quando a condição se torna falsa, o programa continua depois do bloco.

````python
contador = 0

while contador < 5:
    print(contador)
    contador += 1

print('Fim do laço.')
````

### As três partes de um `while`

O exemplo possui três elementos importantes:

1. **inicialização:** `contador = 0`;
2. **condição:** `contador < 5`;
3. **atualização:** `contador += 1`.

A condição é verificada antes de cada repetição. Os valores mostrados são `0`, `1`, `2`, `3` e `4`; quando o contador chega a `5`, a condição fica falsa.

### Acompanhando as iterações

| Iteração | Valor antes do bloco | `contador < 5` | Valor depois de `+= 1` |
|---:|---:|---:|---:|
| 1 | `0` | `True` | `1` |
| 2 | `1` | `True` | `2` |
| 3 | `2` | `True` | `3` |
| 4 | `3` | `True` | `4` |
| 5 | `4` | `True` | `5` |
| parada | `5` | `False` | não executa |

### Contagem regressiva

A atualização também pode diminuir o contador.

````python
contador = 5

while contador > 0:
    print(contador)
    contador -= 1

print('Começou!')
````

## 3. Cuidado com laços infinitos

Se a condição nunca se tornar falsa, o laço não termina. No código abaixo, faltaria atualizar `contador`:

```python
contador = 0
while contador < 5:
    print(contador)
```

Como `contador` continuaria valendo `0`, a condição seria sempre verdadeira. O exemplo não será executado aqui.

> Antes de executar um `while`, identifique o que fará sua condição ficar falsa.

## 4. Repetição com `for`

O `for` percorre os itens de uma sequência. A cada repetição, uma variável recebe o próximo item.

```python
for item in sequencia:
    instruções repetidas
```

````python
frutas = ['banana', 'maçã', 'laranja']

for fruta in frutas:
    print(f'Fruta atual: {fruta}')

print('Todos os itens foram percorridos.')
````

Na primeira iteração, `fruta` recebe `'banana'`; na segunda, `'maçã'`; na terceira, `'laranja'`. Depois do último item, o laço termina automaticamente.

````python
medicoes = [30.2, 55, 20]

for medicao in medicoes:
    print(f'Item da lista: {medicao}')
````

## 5. `while` ou `for`?

| Situação | Estrutura mais comum |
|---|---|
| repetir enquanto uma condição for verdadeira | `while` |
| número de repetições ainda é desconhecido | `while` |
| percorrer os itens de uma lista | `for` |
| repetir uma quantidade conhecida de vezes | `for` com `range()` |

As duas estruturas podem resolver alguns dos mesmos problemas, mas uma delas costuma expressar melhor a intenção.

## 6. Gerando números com `range()`

A função `range()` produz uma sequência de números inteiros, muito usada com `for`. Ela pode receber início, fim e passo:

```python
range(inicio, fim, passo)
```

Assim como no fatiamento, o início entra e o fim fica de fora.

### `range(fim)`

Com um argumento, a sequência começa em `0` e termina antes do valor informado.

````python
for numero in range(5):
    print(numero, end=' ')

print()
````

O argumento `end=' '` faz cada `print()` terminar com um espaço em vez de começar uma nova linha. O `print()` vazio no final apenas pula a linha.

### `range(inicio, fim)`

Com dois argumentos, escolhemos o início e o fim exclusivo.

````python
for numero in range(5, 10):
    print(numero, end=' ')

print()
````

A saída vai de `5` até `9`. O valor `10` marca a parada e não é incluído.

### `range(inicio, fim, passo)`

O passo indica de quanto em quanto a sequência avança.

````python
for numero in range(-4, 5, 2):
    print(numero, end=' ')

print()
````

A sequência começa em `-4`, avança de dois em dois e para antes de `5`: `-4, -2, 0, 2, 4`. O passo não pode ser zero.

### Contagem decrescente com `range()`

Para diminuir os valores, usamos um passo negativo. O início precisa ser maior que o fim.

````python
for numero in range(5, 0, -1):
    print(numero, end=' ')

print('Começou!')
````

## 7. Condições dentro de laços

Podemos usar `if`, `elif` e `else` dentro de um laço. A condição será avaliada novamente em cada iteração.

````python
medicoes = [30.2, 55, 20]

for medicao in medicoes:
    if medicao > 30:
        print(f'Medição acima de 30: {medicao}')
````

Observe os dois níveis de indentação: o `if` pertence ao `for`, e o `print()` pertence ao `if`.

````python
for numero in range(1, 11):
    if numero % 2 == 0:
        print(f'{numero} é par.')
    else:
        print(f'{numero} é ímpar.')
````

## 8. Pulando uma iteração com `continue`

O comando `continue` encerra apenas a iteração atual e passa para a próxima. No exemplo, o número `5` não é mostrado.

````python
for numero in range(1, 11):
    if numero == 5:
        continue

    print(numero, end=' ')

print()
````

Quando `numero == 5`, o `continue` pula o `print()` e o `for` segue com `6`.

> Em um `while`, tenha cuidado: usar `continue` antes de atualizar o contador pode criar um laço infinito.

## 9. Encerrando o laço com `break`

O comando `break` encerra todo o laço imediatamente. Ele é útil quando já encontramos o que procurávamos.

````python
numeros = [3, 8, 12, 15, 20]

for numero in numeros:
    print(f'Verificando {numero}...')
    if numero > 10:
        print(f'Primeiro número maior que 10: {numero}')
        break
````

### `continue` e `break`

| Comando | Efeito |
|---|---|
| `continue` | pula o restante da iteração atual |
| `break` | encerra completamente o laço |

## 10. Contadores e acumuladores

Um **contador** registra quantas vezes algo aconteceu. Um **acumulador** reúne valores, como uma soma. Ambos começam antes do laço e são atualizados dentro dele.

````python
numeros = [4, 7, 10, 13, 16]
quantidade_de_pares = 0
soma_dos_pares = 0

for numero in numeros:
    if numero % 2 == 0:
        quantidade_de_pares += 1
        soma_dos_pares += numero

print(f'Quantidade de pares: {quantidade_de_pares}')
print(f'Soma dos pares: {soma_dos_pares}')
````

### Soma passo a passo

Para somar os números de 1 a 5, o acumulador começa em zero. Em cada iteração, o número atual é acrescentado ao total.

````python
total = 0

for numero in range(1, 6):
    total += numero
    print(f'Após adicionar {numero}, total = {total}')

print(f'Soma final: {total}')
````

## 11. Exemplo completo: FizzBuzz

Para cada número de 1 a 20:

- múltiplos de 3 e 5 mostram `FizzBuzz`;
- apenas múltiplos de 3 mostram `Fizz`;
- apenas múltiplos de 5 mostram `Buzz`;
- os demais mostram o próprio número.

O caso mais específico, múltiplo dos dois números, precisa aparecer primeiro.

````python
for numero in range(1, 21):
    if numero % 3 == 0 and numero % 5 == 0:
        print('FizzBuzz', end=' ')
    elif numero % 3 == 0:
        print('Fizz', end=' ')
    elif numero % 5 == 0:
        print('Buzz', end=' ')
    else:
        print(numero, end=' ')

print()
````

## 12. Exemplo completo: limite de tentativas

O exemplo simula três senhas digitadas. Uma lista é usada no lugar de `input()` para que o notebook execute automaticamente. O `while` limita as tentativas e o `break` encerra o laço quando a senha correta é encontrada.

````python
senha_correta = 'senha123'
senhas_digitadas = ['teste', 'abc123', 'senha123']
tentativas = 0
maximo_de_tentativas = 3
acesso_concedido = False

while tentativas < maximo_de_tentativas:
    senha = senhas_digitadas[tentativas]
    tentativas += 1
    print(f'Tentativa {tentativas}: {senha}')

    if senha == senha_correta:
        acesso_concedido = True
        print('Acesso concedido.')
        break

    print('Senha incorreta.')

if not acesso_concedido:
    print('Acesso bloqueado após três tentativas.')
````

Para transformar a simulação em um programa interativo, a linha que escolhe a senha na lista pode ser substituída por:

```python
senha = input('Digite a senha: ')
```

A atualização de `tentativas` deve continuar dentro do laço.

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `while contador < 10` | faltaram os dois-pontos | use `while contador < 10:` |
| não atualizar a condição do `while` | o laço pode nunca terminar | atualize o contador ou estado |
| esperar que `range(1, 10)` inclua `10` | o fim é exclusivo | use `range(1, 11)` |
| `range(10, 1)` para contar para trás | o passo padrão é positivo | use `range(10, 0, -1)` |
| `range(1, 10, 0)` | o passo não pode ser zero | escolha um passo diferente de zero |
| colocar o acumulador dentro do laço | ele volta ao valor inicial a cada iteração | inicialize antes do laço |
| confundir `continue` com `break` | um pula e o outro encerra | escolha conforme o efeito desejado |
| indentação incorreta no `if` interno | os blocos ficam associados de forma errada | confira cada nível de recuo |
