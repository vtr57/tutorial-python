## 1. O que é uma variável?

Imagine uma caixa com uma etiqueta. A **etiqueta** é o nome da variável e o que está dentro da caixa é o seu **valor**.

Por exemplo, podemos criar uma caixa chamada `idade` e guardar o número `16` dentro dela:

````python
idade = 16
print(idade)
````

Leia `idade = 16` como: **a variável `idade` recebe o valor `16`**.

O símbolo `=` faz uma **atribuição**. Em Python, ele guarda o valor que está à direita no nome que está à esquerda.

> Atenção: aqui, `=` não é uma pergunta matemática. É uma instrução para o computador guardar um valor.

````python
# Uma variável pode receber um novo valor.
pontos = 10
print('Antes:', pontos)

pontos = 15
print('Depois:', pontos)
````

### Escolhendo bons nomes

Um bom nome ajuda qualquer pessoa a entender o programa.

| Prefira | Evite | Motivo |
|---|---|---|
| `nome_aluno` | `x` | o primeiro nome explica o conteúdo |
| `nota_final` | `nf` | abreviações podem causar confusão |
| `temperatura` | `Temperatura` | nomes de variáveis geralmente começam com letra minúscula |

Regras importantes:

- use letras, números e `_`;
- não comece o nome com um número;
- não use espaços; use `_` para separar palavras;
- Python diferencia maiúsculas de minúsculas: `idade` e `Idade` são nomes diferentes.

## 2. Tipos de dados

O tipo indica qual espécie de informação está guardada. Nesta aula, vamos estudar três tipos básicos:

| Tipo | O que guarda | Exemplos |
|---|---|---|
| `int` | números inteiros | `-5`, `0`, `16`, `2026` |
| `float` | números com parte decimal | `3.14`, `9.99`, `25.0` |
| `str` | textos | `'Ana'`, `'2º ano'`, `'11999999999'` |

O Python identifica o tipo automaticamente a partir do valor informado.

## 3. Números inteiros: `int`

O tipo `int` representa números inteiros, isto é, números sem parte decimal. Eles podem ser negativos, positivos ou zero.

````python
temperatura_negativa = -5
saldo_de_gols = 0
total_de_alunos = 35

print(temperatura_negativa)
print(saldo_de_gols)
print(total_de_alunos)
````

### Como descobrir um tipo

A função `type()` pergunta ao Python qual é o tipo de um valor. O resultado `<class 'int'>` significa que o valor pertence ao tipo `int`.

````python
ano_atual = 2026
print(ano_atual)
print(type(ano_atual))
````

### Pare e pense

Sem executar código, qual seria o tipo de cada valor?

- `12`
- `-8`
- `0`

Todos são `int`, pois nenhum possui parte decimal.

## 4. Números decimais: `float`

O tipo `float` representa números com parte decimal. Ele é útil para preços, medidas, notas e temperaturas.

> Em Python, usamos **ponto** como separador decimal: escreva `9.99`, e não `9,99`.

````python
preco_lanche = 9.99
altura = 1.72
nota = 8.5
temperatura = 25.0

print(preco_lanche)
print(altura)
print(nota)
print(temperatura)
print(type(preco_lanche))
````

### Uma forma diferente de escrever números

Python também entende notação científica. Por exemplo, `1e-4` significa $1 \times 10^{-4}$, ou seja, `0.0001`. Esse recurso aparece bastante em Ciências da Natureza e Matemática.

````python
numero_pequeno = 1e-4
print(numero_pequeno)
print(type(numero_pequeno))
````

## 5. Textos: `str`

O tipo `str` representa uma sequência de caracteres: letras, números, espaços e símbolos. Em Python, textos devem ficar entre aspas simples ou duplas.

A palavra `str` vem de *string*, que pode ser entendida como uma sequência de caracteres.

````python
nome = 'Vitor'
cargo = "Professor"
turma = '2º ano B'
cpf = '77788899900'

print(nome)
print(cargo)
print(turma)
print(cpf)
print(type(nome))
````

### Por que um CPF pode ser `str`?

Embora um CPF seja formado por algarismos, normalmente não fazemos contas com ele. Ele funciona como uma identificação e pode até começar com zero. Por isso, faz mais sentido guardá-lo como texto.

A mesma ideia vale para número de telefone, CEP, matrícula e código de um produto.

````python
# Os algarismos estão entre aspas, então o valor é um texto.
cep = '01234-567'
telefone = '11987654321'

print(cep, type(cep))
print(telefone, type(telefone))
````

## 6. Mostrando informações com `print()`

A função `print()` exibe informações na tela. Ela pode mostrar um texto, uma variável ou vários valores separados por vírgulas.

````python
nome_aluno = 'Mariana'
idade_aluno = 16

print('Dados do estudante:')
print(nome_aluno)
print('Nome:', nome_aluno, '- Idade:', idade_aluno)
````

## 7. Criando mensagens com f-strings

Uma f-string facilita a mistura de texto e variáveis. Colocamos a letra `f` antes das aspas e o nome de cada variável entre chaves `{}`.

Formato geral: `f'Texto {variavel}'`

````python
nome_aluno = 'Mariana'
idade_aluno = 16
temperatura = 25.0

print(f'Meu nome é {nome_aluno} e tenho {idade_aluno} anos.')
print(f'A temperatura de hoje é {temperatura} graus Celsius.')
````

### Erros comuns no começo

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `nome = Ana` | faltaram aspas no texto | `nome = 'Ana'` |
| `preco = 9,99` | foi usada vírgula decimal | `preco = 9.99` |
| `2idade = 16` | o nome começou com número | `idade2 = 16` |
| `nome aluno = 'Ana'` | o nome contém espaço | `nome_aluno = 'Ana'` |
| `print(f'Olá, nome')` | faltaram chaves na variável | `print(f'Olá, {nome}')` |

Mensagens de erro fazem parte da programação. Leia a última linha da mensagem, localize a célula e confira aspas, nomes e sinais.

## 8. Exemplo completo

Antes de executar, tente prever as quatro linhas que aparecerão.

````python
nome = 'Lucas'
idade = 17
altura = 1.75
turma = '3º ano A'

print(f'Nome: {nome}')
print(f'Idade: {idade} anos')
print(f'Altura: {altura} m')
print(f'Turma: {turma}')
````
