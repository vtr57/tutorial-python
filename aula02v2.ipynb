## 2. Operadores aritméticos

Em Python, cada operação é representada por um símbolo chamado **operador**.

| Operação | Operador | Exemplo | Resultado |
|---|---:|---:|---:|
| adição | `+` | `8 + 2` | `10` |
| subtração | `-` | `8 - 2` | `6` |
| multiplicação | `*` | `8 * 2` | `16` |

## 3. Adição: `+`

A adição combina duas quantidades. Os valores usados na operação são chamados de **operandos**, e a resposta pode ser guardada em outra variável.

````python
a = 237
b = 1259
resultado = a + b

print(f'{a} + {b} = {resultado}')
````

A operação também pode misturar `int` e `float`. Nesse caso, o resultado normalmente será `float`.

````python
valor_inteiro = 10
valor_decimal = 2.5
soma = valor_inteiro + valor_decimal

print(f'{valor_inteiro} + {valor_decimal} = {soma}')
print(type(soma))
````

## 6. Divisão comum: `/`

Em Python, o operador `/` produz um resultado do tipo `float`, mesmo quando a divisão é exata.

````python
total = 20
numero_de_pessoas = 4
parte_de_cada_pessoa = total / numero_de_pessoas

print(f'{total} / {numero_de_pessoas} = {parte_de_cada_pessoa}')
print(type(parte_de_cada_pessoa))
````

## 7. Divisão inteira `//` e módulo `%`

Imagine 17 estudantes formando grupos de 5:

- `17 // 5` informa quantos grupos completos podem ser formados;
- `17 % 5` informa quantos estudantes ficam sem grupo completo.

````python
estudantes = 17
tamanho_do_grupo = 5
grupos_completos = estudantes // tamanho_do_grupo
estudantes_restantes = estudantes % tamanho_do_grupo

print(f'Grupos completos: {grupos_completos}')
print(f'Estudantes restantes: {estudantes_restantes}')
````

A divisão inteira e o resto obedecem à relação:

$$	dividendo = 	divisor \times 	{quociente} + 	{resto}$$

No exemplo: $17 = 5 \times 3 + 2$.

````python
dividendo = 17
divisor = 5
quociente = dividendo // divisor
resto = dividendo % divisor
reconstrucao = divisor * quociente + resto

print(f'{dividendo} = {divisor} * {quociente} + {resto}')
print(f'Resultado da reconstrução: {reconstrucao}')
````

### Verificando divisibilidade

Um número é divisível por outro quando o resto da divisão é zero.

- Se `numero % divisor` for `0`, a divisão é exata.
- Se o resto for diferente de `0`, a divisão não é exata.

````python
numero = 120
divisor = 10
resto = numero % divisor

print(f'O resto de {numero} dividido por {divisor} é {resto}.')
print('Como o resto é zero, 120 é divisível por 10.')
````

O resto também ajuda a reconhecer números pares e ímpares. Todo número par deixa resto `0` quando dividido por `2`; todo número ímpar deixa resto `1`.

````python
numero_par = 28
numero_impar = 35

print(f'{numero_par} % 2 = {numero_par % 2}')
print(f'{numero_impar} % 2 = {numero_impar % 2}')
````

## 8. Potenciação: `**`

Uma potência representa uma multiplicação repetida. Em $3^4$, o número `3` é a base e `4` é o expoente.

Em Python, escrevemos `3 ** 4`, e não `3 ^ 4`.

````python
base = 3
expoente = 4
potencia = base ** expoente

print(f'{base} elevado a {expoente} é {potencia}.')
````

### Cuidado com o símbolo `^`

Na Matemática, costumamos escrever $3^2$. Em Python, porém, `^` possui outra função e **não calcula potência**. Para calcular três ao quadrado, use `3 ** 2`.

````python
lado_do_quadrado = 6
area = lado_do_quadrado ** 2

print(f'A área de um quadrado de lado {lado_do_quadrado} é {area}.')
````

## 9. Ordem das operações

Python segue uma ordem parecida com a usada na Matemática:

1. parênteses `()`;
2. potências `**`;
3. multiplicações e divisões `*`, `/`, `//`, `%`;
4. adições e subtrações `+`, `-`.

Operações de mesma prioridade são geralmente resolvidas da esquerda para a direita. Use parênteses quando quiser deixar a intenção mais clara.

````python
sem_parenteses = 2 + 3 * 4
com_parenteses = (2 + 3) * 4

print(f'2 + 3 * 4 = {sem_parenteses}')
print(f'(2 + 3) * 4 = {com_parenteses}')
````

### Exemplo passo a passo

Vamos calcular $(3^2 + 10) \times (25 - 4) \div 6$. Em Python, a expressão é:

```python
(3 ** 2 + 10) * (25 - 4) / 6
```

A ordem será: potência, parênteses, multiplicação e divisão.

````python
potencia = 3 ** 2
primeiro_parenteses = potencia + 10
segundo_parenteses = 25 - 4
produto = primeiro_parenteses * segundo_parenteses
resultado = produto / 6

print(f'3 ** 2 = {potencia}')
print(f'3 ** 2 + 10 = {primeiro_parenteses}')
print(f'25 - 4 = {segundo_parenteses}')
print(f'{primeiro_parenteses} * {segundo_parenteses} = {produto}')
print(f'{produto} / 6 = {resultado}')
````

O mesmo cálculo pode ser escrito em uma linha quando já entendemos suas etapas.

````python
resultado = (3 ** 2 + 10) * (25 - 4) / 6
print(resultado)
````

## 10. Atualizando o valor de uma variável

Às vezes queremos usar o valor atual de uma variável para calcular seu próximo valor.

```python
pontos = pontos + 10
```

Podemos abreviar essa instrução como `pontos += 10`.

````python
pontos = 100
print(f'Pontuação inicial: {pontos}')

pontos += 30
print(f'Depois de ganhar 30 pontos: {pontos}')

pontos -= 20
print(f'Depois de perder 20 pontos: {pontos}')
````

Há versões abreviadas para várias operações:

| Forma abreviada | Equivale a |
|---|---|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |

> Essas instruções alteram o valor guardado na variável.

````python
valor = 10
print(f'Início: {valor}')

valor *= 2
print(f'Após valor *= 2: {valor}')

valor /= 4
print(f'Após valor /= 4: {valor}')
````

## 11. Resultados decimais e `round()`

Alguns cálculos produzem muitas casas decimais. A função `round(valor, casas)` cria um valor arredondado para a quantidade de casas indicada.

````python
media = 20 / 3
media_arredondada = round(media, 2)

print(f'Resultado completo: {media}')
print(f'Resultado com duas casas decimais: {media_arredondada}')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `3 ^ 2` para calcular potência | `^` não é potência em Python | `3 ** 2` |
| `10 / 0` | não existe divisão por zero | use um divisor diferente de zero |
| `2(3 + 4)` | faltou o operador de multiplicação | `2 * (3 + 4)` |
| `preco = 9,99` | foi usada vírgula decimal | `preco = 9.99` |
| `resultado = a + b` sem criar `a` e `b` | as variáveis não existem ainda | atribua valores antes do cálculo |

Se aparecer um erro, leia a última linha da mensagem e confira operadores, parênteses e nomes de variáveis.

## 12. Exemplo completo: calculando uma compra

Uma pessoa comprou quatro cadernos de R$ 12,50 e recebeu R$ 8,00 de desconto. Vamos calcular o valor final e dividir a conta entre duas pessoas.

````python
quantidade = 4
preco_unitario = 12.50
desconto = 8.00
numero_de_pessoas = 2

subtotal = quantidade * preco_unitario
total = subtotal - desconto
valor_por_pessoa = total / numero_de_pessoas

print(f'Subtotal: R$ {subtotal}')
print(f'Desconto: R$ {desconto}')
print(f'Total: R$ {total}')
print(f'Cada pessoa pagará R$ {valor_por_pessoa}.')
````
