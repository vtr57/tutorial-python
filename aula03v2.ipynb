## 1. Perguntas que só têm duas respostas

Algumas perguntas podem ser respondidas apenas com **sim** ou **não**:

- 10 é maior que 7? Sim.
- 4 é igual a 9? Não.
- Hoje é o ano de 2030? Depende do ano atual, mas a resposta ainda será sim ou não.

Em Python, usamos `True` para verdadeiro e `False` para falso. Esses valores pertencem ao tipo `bool`, chamado de tipo booleano.

````python
print(True)
print(False)
print(type(True))
print(type(False))
````

### Atenção à escrita

Python diferencia letras maiúsculas de minúsculas. Por isso, escreva `True` e `False` com a primeira letra maiúscula.

Também não use aspas quando quiser representar um valor booleano:

- `True` é um valor do tipo `bool`;
- `'True'` é um texto do tipo `str`.

````python
valor_booleano = True
texto = 'True'

print(valor_booleano, type(valor_booleano))
print(texto, type(texto))
````

Uma variável também pode guardar o resultado de uma pergunta lógica.

````python
idade = 16
maior_de_idade = idade >= 18

print(f'Idade: {idade}')
print(f'É maior de idade? {maior_de_idade}')
print(type(maior_de_idade))
````

## 2. Operadores de comparação

Os operadores de comparação fazem uma pergunta sobre dois valores. O resultado sempre será `True` ou `False`.

| Operador | Pergunta | Exemplo | Resultado |
|---:|---|---:|---:|
| `>` | é maior que? | `10 > 9` | `True` |
| `<` | é menor que? | `10 < 9` | `False` |
| `>=` | é maior ou igual a? | `10 >= 10` | `True` |
| `<=` | é menor ou igual a? | `7 <= 5` | `False` |
| `==` | é igual a? | `10 == 10` | `True` |
| `!=` | é diferente de? | `10 != 9` | `True` |

### Maior e menor: `>` e `<`

A abertura do símbolo fica voltada para o maior valor. Leia `10 > 9` como “dez é maior que nove?”.

````python
print(f'10 > 9: {10 > 9}')
print(f'10 < 9: {10 < 9}')
print(f'-2 > -5: {-2 > -5}')
print(f'3.5 < 4: {3.5 < 4}')
````

### Maior ou igual e menor ou igual: `>=` e `<=`

Esses operadores aceitam duas possibilidades. `nota >= 6` será verdadeiro quando a nota for maior que 6 **ou exatamente igual a 6**.

````python
nota = 6.0
print(f'A nota {nota} é maior ou igual a 6? {nota >= 6}')
print(f'A nota {nota} é menor ou igual a 5? {nota <= 5}')
````

### Igual e diferente: `==` e `!=`

O operador `==` pergunta se dois valores são iguais. O operador `!=` pergunta se eles são diferentes.

````python
print(f'10 == 9: {10 == 9}')
print(f'10 != 9: {10 != 9}')
print(f'10 == 5 * 2: {10 == 5 * 2}')
print(f'10 != 3 * 3: {10 != 3 * 3}')
````

### `=` não é a mesma coisa que `==`

- `=` faz uma **atribuição**: guarda um valor em uma variável.
- `==` faz uma **comparação**: pergunta se dois valores são iguais.

Leia `nota = 8` como “nota recebe 8” e `nota == 8` como “nota é igual a 8?”.

````python
nota = 8
comparacao = nota == 8

print(f'Valor guardado em nota: {nota}')
print(f'Nota é igual a 8? {comparacao}')
````

### Comparando textos

Textos também podem ser comparados com `==` e `!=`. A comparação considera todas as letras, inclusive maiúsculas, minúsculas e acentos.

````python
turma = '2º ano A'

print(f"turma == '2º ano A': {turma == '2º ano A'}")
print(f"turma == '2º ano B': {turma == '2º ano B'}")
print(f"'Python' == 'python': {'Python' == 'python'}")
````

## 3. Negação: `not`

O operador `not` inverte um valor lógico:

| Expressão | Resultado |
|---|---|
| `not True` | `False` |
| `not False` | `True` |

Podemos entender `not` como a palavra “não”.

````python
print(f'not True = {not True}')
print(f'not False = {not False}')
````

O `not` também pode negar o resultado de uma comparação.

````python
chovendo = False
nao_esta_chovendo = not chovendo

print(f'Está chovendo? {chovendo}')
print(f'Não está chovendo? {nao_esta_chovendo}')
````

## 4. Conjunção: `and`

O operador `and` significa **e**. O resultado só é verdadeiro quando as duas condições são verdadeiras.

| A | B | `A and B` |
|---|---|---|
| `True` | `True` | `True` |
| `True` | `False` | `False` |
| `False` | `True` | `False` |
| `False` | `False` | `False` |

Pense em uma porta que exige **cartão válido e senha correta**: as duas condições precisam ser atendidas.

````python
print(f'True and True = {True and True}')
print(f'True and False = {True and False}')
print(f'False and True = {False and True}')
print(f'False and False = {False and False}')
````

````python
cartao_valido = True
senha_correta = False
acesso_liberado = cartao_valido and senha_correta

print(f'Cartão válido? {cartao_valido}')
print(f'Senha correta? {senha_correta}')
print(f'Acesso liberado? {acesso_liberado}')
````

## 5. Disjunção: `or`

O operador `or` significa **ou**. O resultado é verdadeiro quando pelo menos uma condição é verdadeira.

| A | B | `A or B` |
|---|---|---|
| `True` | `True` | `True` |
| `True` | `False` | `True` |
| `False` | `True` | `True` |
| `False` | `False` | `False` |

Pense em uma entrada que aceita **carteirinha ou ingresso**: basta apresentar um dos dois.

````python
print(f'True or True = {True or True}')
print(f'True or False = {True or False}')
print(f'False or True = {False or True}')
print(f'False or False = {False or False}')
````

````python
tem_carteirinha = False
tem_ingresso = True
pode_entrar = tem_carteirinha or tem_ingresso

print(f'Tem carteirinha? {tem_carteirinha}')
print(f'Tem ingresso? {tem_ingresso}')
print(f'Pode entrar? {pode_entrar}')
````

### `and` ou `or`?

Faça esta pergunta:

- **Todas** as condições precisam ser verdadeiras? Use `and`.
- **Pelo menos uma** condição precisa ser verdadeira? Use `or`.

## 6. Combinando comparações

Normalmente usamos `and` e `or` para combinar perguntas feitas com operadores de comparação.

No exemplo abaixo, uma pessoa pode participar se tiver idade entre 14 e 18 anos, incluindo os limites.

````python
idade = 16
tem_idade_minima = idade >= 14
tem_idade_maxima = idade <= 18
pode_participar = tem_idade_minima and tem_idade_maxima

print(f'Tem pelo menos 14 anos? {tem_idade_minima}')
print(f'Tem no máximo 18 anos? {tem_idade_maxima}')
print(f'Pode participar? {pode_participar}')
````

A mesma expressão pode ser escrita diretamente, sem variáveis intermediárias. As variáveis intermediárias, porém, costumam facilitar a leitura no início.

````python
idade = 16
pode_participar = idade >= 14 and idade <= 18
print(pode_participar)
````

## 7. Ordem das operações lógicas

Quando uma expressão possui vários operadores, Python segue esta ordem simplificada:

1. parênteses `()`;
2. operações matemáticas;
3. comparações como `>`, `<`, `==` e `!=`;
4. `not`;
5. `and`;
6. `or`.

Use parênteses para deixar claro quais partes devem ser avaliadas juntas, mesmo quando você conhece essa ordem.

### Exemplo passo a passo

Vamos avaliar:

```python
10 == 5 * 2 and 10 != 3 * 3
```

Primeiro vêm as multiplicações, depois as comparações e, por último, o `and`.

````python
primeira_multiplicacao = 5 * 2
segunda_multiplicacao = 3 * 3
primeira_comparacao = 10 == primeira_multiplicacao
segunda_comparacao = 10 != segunda_multiplicacao
resultado = primeira_comparacao and segunda_comparacao

print(f'5 * 2 = {primeira_multiplicacao}')
print(f'3 * 3 = {segunda_multiplicacao}')
print(f'10 == 10: {primeira_comparacao}')
print(f'10 != 9: {segunda_comparacao}')
print(f'True and True: {resultado}')
````

### Os parênteses podem mudar o resultado

Compare `not True and False` com `not (True and False)`. Na primeira expressão, apenas `True` é negado. Na segunda, o resultado inteiro de `True and False` é negado.

````python
sem_parenteses = not True and False
com_parenteses = not (True and False)

print(f'not True and False = {sem_parenteses}')
print(f'not (True and False) = {com_parenteses}')
````

## 8. Exemplo completo: participação em uma atividade

Para participar de uma atividade, o estudante precisa ter média maior ou igual a 6 **e** frequência maior ou igual a 75%. Além disso, a inscrição não pode estar bloqueada.

````python
media = 7.5
frequencia = 82
inscricao_bloqueada = False

media_suficiente = media >= 6
frequencia_suficiente = frequencia >= 75
inscricao_liberada = not inscricao_bloqueada
pode_participar = media_suficiente and frequencia_suficiente and inscricao_liberada

print(f'Média suficiente? {media_suficiente}')
print(f'Frequência suficiente? {frequencia_suficiente}')
print(f'Inscrição liberada? {inscricao_liberada}')
print(f'Pode participar? {pode_participar}')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `nota = 7` quando a intenção era comparar | `=` faz atribuição | use `nota == 7` |
| `true` ou `false` | booleanos começam com maiúscula | use `True` ou `False` |
| `'True'` | as aspas criam um texto | use `True` sem aspas |
| `idade > 14 and < 18` | a segunda comparação está incompleta | use `idade > 14 and idade < 18` |
| `nota => 6` | os símbolos estão invertidos | use `nota >= 6` |
| uma expressão longa sem parênteses | a intenção pode ficar confusa | agrupe as condições relacionadas |

Ao encontrar um erro, confira maiúsculas, operadores e se cada comparação possui valores nos dois lados.
