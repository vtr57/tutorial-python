## 1. Por que usar herança?

Aves e répteis possuem diferenças, mas ambos são animais. Podemos colocar dados e comportamentos comuns em `Animal` e deixar cada subclasse acrescentar suas características.

Sem herança, seria necessário repetir `peso`, `idade`, `especie`, `acordar()`, `dormir()` e outros métodos em várias classes.

### Vocabulário

| Termo | Outro nome | Exemplo |
|---|---|---|
| classe base | classe pai ou superclasse | `Animal` |
| subclasse | classe filha ou classe derivada | `Ave` |
| herdar | receber recursos da classe base | `Ave` recebe `dormir()` |
| sobrescrever | fornecer nova versão de um método | `Cachorro.emitir_som()` |

## 2. Criando a classe base

A classe base reúne o que é comum às subclasses.

````python
class Animal:
    def __init__(self, peso, idade, especie):
        self.peso = peso
        self.idade = idade
        self.especie = especie

    def acordar(self):
        return f'{self.especie} acordando'

    def dormir(self):
        return f'{self.especie} dormindo'

    def correr(self):
        return f'{self.especie} correndo'

    def comer(self):
        return f'{self.especie} comendo'

    def emitir_som(self):
        return f'{self.especie} emitindo um som'
````

`Animal` já pode criar objetos diretamente, mas seu principal papel aqui será fornecer a estrutura compartilhada.

````python
animal = Animal(peso=12, idade=3, especie='Animal genérico')

print(animal.acordar())
print(animal.comer())
print(animal.dormir())
````

## 3. Criando uma subclasse

Colocamos a classe base entre parênteses após o nome da subclasse:

```python
class Subclasse(ClasseBase):
    ...
```

````python
class Reptil(Animal):
    def rastejar(self):
        return f'{self.especie} rastejando'
````

`Reptil` não define `__init__`. Python usa automaticamente o inicializador herdado de `Animal`. A subclasse também recebe todos os métodos da base e acrescenta `rastejar()`.

````python
lagarto = Reptil(peso=4, idade=2, especie='Lagarto')

print(lagarto.acordar())
print(lagarto.comer())
print(lagarto.correr())
print(lagarto.rastejar())
print(lagarto.dormir())
````

`acordar`, `comer`, `correr` e `dormir` são herdados. `rastejar` pertence especificamente a `Reptil`.

## 4. Quando herdar o `__init__` sem reescrever

Se a subclasse precisa exatamente dos mesmos atributos iniciais, não é necessário repetir o inicializador.

````python
class Peixe(Animal):
    def nadar(self):
        return f'{self.especie} nadando'

peixe = Peixe(2, 1, 'Peixe-palhaço')

print(f'Peso herdado: {peixe.peso} kg')
print(peixe.nadar())
print(peixe.dormir())
````

Repetir um `__init__` que apenas encaminha os mesmos argumentos adiciona código sem mudar o comportamento.

## 5. Acrescentando atributos com `super()`

Quando a subclasse precisa de um atributo novo, podemos definir seu próprio `__init__`. `super().__init__()` chama o inicializador da classe base para configurar a parte comum.

````python
class Ave(Animal):
    def __init__(self, peso, idade, especie, envergadura):
        super().__init__(peso, idade, especie)
        self.envergadura = envergadura

    def voar(self):
        return f'{self.especie} voando'
````

A linha com `super()` cria `peso`, `idade` e `especie`. A linha seguinte cria somente o atributo específico `envergadura`.

````python
arara = Ave(peso=1.2, idade=4, especie='Arara', envergadura=1.1)

print(arara.acordar())
print(arara.comer())
print(arara.voar())
print(f'Envergadura: {arara.envergadura} m')
print(arara.dormir())
````

### O que acontece sem `super()`?

Se o `__init__` da subclasse não chamar o da base e também não criar os atributos comuns, nomes como `self.especie` não existirão. Métodos herdados que dependem deles poderão provocar `AttributeError`.

## 6. Sobrescrevendo um método

Uma subclasse pode definir um método com o mesmo nome da classe base. A nova versão será usada por objetos dessa subclasse.

````python
class Cachorro(Animal):
    def emitir_som(self):
        return f'{self.especie}: au au!'

class Gato(Animal):
    def emitir_som(self):
        return f'{self.especie}: miau!'

cachorro = Cachorro(18, 5, 'Cachorro')
gato = Gato(4, 3, 'Gato')

print(animal.emitir_som())
print(cachorro.emitir_som())
print(gato.emitir_som())
````

`Animal` mantém a implementação genérica. `Cachorro` e `Gato` fornecem versões específicas. Os outros métodos continuam herdados normalmente.

````python
print(cachorro.comer())
print(gato.dormir())
````

## 7. Objetos diferentes, mesma chamada

Como todas as classes possuem `emitir_som()`, podemos percorrer objetos diferentes e realizar a mesma chamada. Cada objeto usa sua própria implementação.

````python
animais = [animal, cachorro, gato]

for animal_atual in animais:
    print(animal_atual.emitir_som())
````

Esse comportamento, no qual a mesma operação assume formas diferentes conforme o objeto, é uma introdução à ideia de **polimorfismo**.

## 8. Verificando relações de herança

`isinstance()` verifica a relação de um objeto com uma classe. Um objeto de uma subclasse também é considerado instância de sua classe base.

````python
print(f'arara é Ave? {isinstance(arara, Ave)}')
print(f'arara é Animal? {isinstance(arara, Animal)}')
print(f'lagarto é Reptil? {isinstance(lagarto, Reptil)}')
print(f'lagarto é Ave? {isinstance(lagarto, Ave)}')
````

`issubclass()` verifica a relação entre duas classes.

````python
print(f'Ave é subclasse de Animal? {issubclass(Ave, Animal)}')
print(f'Reptil é subclasse de Animal? {issubclass(Reptil, Animal)}')
print(f'Animal é subclasse de Ave? {issubclass(Animal, Ave)}')
````

## 9. Relação “é um”

Herança costuma fazer sentido quando podemos dizer que a subclasse **é um tipo de** classe base:

- uma ave **é um** animal;
- um carro **é um** veículo;
- um estudante **é uma** pessoa.

Ela não é adequada apenas porque um objeto usa outro:

- um motor não é um carro; o carro **possui** um motor;
- um livro não é uma biblioteca; a biblioteca **possui** livros.

Nesses casos de “possui”, normalmente usamos atributos em vez de herança.

## 10. Exemplo: veículos

A classe base guarda marca e modelo. As subclasses acrescentam dados próprios.

````python
class Veiculo:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo

    def descrever(self):
        return f'{self.marca} {self.modelo}'

class Carro(Veiculo):
    def __init__(self, marca, modelo, numero_de_portas):
        super().__init__(marca, modelo)
        self.numero_de_portas = numero_de_portas

    def descrever(self):
        descricao_base = super().descrever()
        return f'{descricao_base} - {self.numero_de_portas} portas'

class Moto(Veiculo):
    def __init__(self, marca, modelo, tipo_motor):
        super().__init__(marca, modelo)
        self.tipo_motor = tipo_motor

    def descrever(self):
        descricao_base = super().descrever()
        return f'{descricao_base} - motor {self.tipo_motor}'
````

Além de chamar `super().__init__()`, um método sobrescrito pode chamar a implementação da base com `super().descrever()` e complementar seu resultado.

````python
carro = Carro('Toyota', 'Corolla', 4)
moto = Moto('Honda', 'CG', '160 cc')

print(carro.descrever())
print(moto.descrever())
````

## 11. Exemplo: pessoas

Estudantes e professores compartilham nome e idade, mas possuem dados específicos.

````python
class Pessoa:
    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

    def apresentar(self):
        return f'{self.nome}, {self.idade} anos'

class Estudante(Pessoa):
    def __init__(self, nome, idade, matricula):
        super().__init__(nome, idade)
        self.matricula = matricula

    def apresentar(self):
        return f'{super().apresentar()} - matrícula {self.matricula}'

class Professor(Pessoa):
    def __init__(self, nome, idade, disciplina):
        super().__init__(nome, idade)
        self.disciplina = disciplina

    def apresentar(self):
        return f'{super().apresentar()} - professor(a) de {self.disciplina}'
````

````python
pessoas = [
    Estudante('Ana', 16, '2026-001'),
    Professor('Vitor', 35, 'Python')
]

for pessoa in pessoas:
    print(pessoa.apresentar())
````

A lista reúne subclasses diferentes. A mesma chamada `apresentar()` produz uma descrição adequada a cada objeto.

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `class Ave:` esperando herança | a classe base não foi informada | use `class Ave(Animal):` |
| repetir `__init__` sem acrescentar nada | código duplicado sem benefício | herde o inicializador |
| criar `__init__` novo e esquecer `super()` | atributos da base podem não existir | chame `super().__init__(...)` |
| passar `self` ao `super().__init__()` | Python já cuida do objeto | envie apenas os demais argumentos |
| sobrescrever um método e esperar execução da versão base | a nova versão substitui a herdada | chame `super().metodo()` se necessário |
| usar herança para uma relação “possui” | a hierarquia fica conceitualmente errada | use um atributo |
| criar muitos níveis sem necessidade | o fluxo fica difícil de seguir | mantenha a hierarquia simples |

## 12. Prática guiada

Altere os dados, acrescente um método específico e crie outra subclasse.

````python
class Instrumento:
    def __init__(self, nome):
        self.nome = nome

    def tocar(self):
        return f'{self.nome} produzindo som'

class Violao(Instrumento):
    def __init__(self, nome, numero_de_cordas):
        super().__init__(nome)
        self.numero_de_cordas = numero_de_cordas

    def tocar(self):
        return f'{self.nome} tocando com {self.numero_de_cordas} cordas'

violao = Violao('Violão clássico', 6)
print(violao.tocar())
````

## 14. Resumo da aula

- Herança permite reutilizar uma classe base em subclasses.
- A sintaxe é `class Subclasse(ClasseBase):`.
- Uma subclasse herda atributos e métodos.
- Se não houver novos atributos, o `__init__` pode ser herdado.
- `super().__init__()` inicializa a parte definida na base.
- Uma subclasse pode acrescentar ou sobrescrever métodos.
- `super().metodo()` permite aproveitar uma implementação herdada.
- Objetos de subclasses também são instâncias da classe base.
- Herança representa uma relação “é um”; atributos representam “possui”.

### Gabarito do Nível 2

````python
class ContaBancaria:
    def __init__(self, titular, saldo):
        self.titular = titular
        self.saldo = saldo

    def consultar_limite(self):
        return self.saldo

class ContaCorrente(ContaBancaria):
    def __init__(self, titular, saldo, limite_cheque_especial):
        super().__init__(titular, saldo)
        self.limite_cheque_especial = limite_cheque_especial

    def consultar_limite(self):
        return self.saldo + self.limite_cheque_especial

class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def descrever(self):
        return f'{self.nome}: R$ {self.preco:.2f}'

class Eletronico(Produto):
    def __init__(self, nome, preco, garantia_meses):
        super().__init__(nome, preco)
        self.garantia_meses = garantia_meses

    def descrever(self):
        return f'{super().descrever()} - garantia de {self.garantia_meses} meses'

class Roupa(Produto):
    def __init__(self, nome, preco, tamanho):
        super().__init__(nome, preco)
        self.tamanho = tamanho

    def descrever(self):
        return f'{super().descrever()} - tamanho {self.tamanho}'

conta = ContaCorrente('Ana', 100, 200)
assert conta.consultar_limite() == 300

produtos = [
    Eletronico('Fone', 150, 12),
    Roupa('Camiseta', 60, 'M')
]
for produto in produtos:
    print(produto.descrever())
````

### Uma solução para o desafio

````python
class ItemBiblioteca:
    def __init__(self, titulo):
        self.titulo = titulo
        self.disponivel = True

    def emprestar(self):
        if not self.disponivel:
            return False
        self.disponivel = False
        return True

    def devolver(self):
        self.disponivel = True

    def descrever(self):
        status = 'disponível' if self.disponivel else 'emprestado'
        return f'{self.titulo} - {status}'

class Livro(ItemBiblioteca):
    def __init__(self, titulo, autor):
        super().__init__(titulo)
        self.autor = autor

    def descrever(self):
        return f'{super().descrever()} - autor: {self.autor}'

class Revista(ItemBiblioteca):
    def __init__(self, titulo, edicao):
        super().__init__(titulo)
        self.edicao = edicao

    def descrever(self):
        return f'{super().descrever()} - edição: {self.edicao}'

class MidiaDigital(ItemBiblioteca):
    def __init__(self, titulo, formato):
        super().__init__(titulo)
        self.formato = formato

    def descrever(self):
        return f'{super().descrever()} - formato: {self.formato}'

itens = [
    Livro('Quarto de Despejo', 'Carolina Maria de Jesus'),
    Revista('Ciência Hoje', 410),
    MidiaDigital('Curso de Python', 'MP4')
]

for item in itens:
    print(item.descrever())

print(f'Primeiro empréstimo: {itens[0].emprestar()}')
print(f'Segundo empréstimo: {itens[0].emprestar()}')
print(itens[0].descrever())
itens[0].devolver()
print(itens[0].descrever())
````
