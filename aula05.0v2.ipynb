## 1. O que é uma estrutura condicional?

Uma estrutura condicional permite que o programa escolha o que fazer. No cotidiano, usamos decisões desse tipo o tempo todo:

- **se** estiver chovendo, leve um guarda-chuva;
- **senão**, leve apenas a mochila.

Em Python, escrevemos **se** como `if` e **senão** como `else`.

## 2. Uma decisão simples com `if`

A estrutura básica é:

```python
if condição:
    instrução
```

A instrução indentada só é executada quando a condição resulta em `True`.

````python
esta_chovendo = True

if esta_chovendo:
    print('Leve um guarda-chuva.')

print('Esta mensagem está fora do if e sempre será executada.')
````

Se a condição for falsa, o bloco indentado é ignorado e o programa continua depois dele.

````python
esta_chovendo = False

if esta_chovendo:
    print('Leve um guarda-chuva.')

print('O programa continuou.')
````

Normalmente, a condição é uma comparação.

````python
temperatura = 31

if temperatura > 30:
    print('O dia está muito quente.')
````

## 3. Dois-pontos e indentação

Duas partes da sintaxe são essenciais:

- os dois-pontos `:` aparecem no final da linha do `if`;
- as instruções pertencentes ao `if` ficam deslocadas para a direita. Esse deslocamento é chamado de **indentação**.

Por convenção, usamos quatro espaços para cada nível de indentação.

````python
nota = 8.0

if nota >= 7:
    print('A condição é verdadeira.')
    print('As duas mensagens pertencem ao bloco do if.')

print('Esta mensagem não pertence ao bloco.')
````

A indentação não é apenas estética: ela informa ao Python quais instruções pertencem a cada caminho. Uma indentação ausente ou inconsistente pode causar `IndentationError`.

## 4. Dois caminhos com `if` e `else`

O `else` representa o caminho alternativo. Exatamente um dos dois blocos será executado:

```python
if condição:
    caminho quando a condição é verdadeira
else:
    caminho quando a condição é falsa
```

O `else` não recebe condição. Ele significa “em qualquer outro caso”.

````python
condicao = True

if condicao:
    print('A condição é verdadeira: bloco do if.')
else:
    print('A condição é falsa: bloco do else.')
````

````python
condicao = False

if condicao:
    print('A condição é verdadeira: bloco do if.')
else:
    print('A condição é falsa: bloco do else.')
````

### Exemplo: número par ou ímpar

Um número é par quando o resto de sua divisão por 2 é zero. Caso contrário, é ímpar.

````python
numero = 17

if numero % 2 == 0:
    print(f'{numero} é par.')
else:
    print(f'{numero} é ímpar.')
````

## 5. Vários caminhos com `elif`

Quando há mais de duas possibilidades, usamos `elif`, uma abreviação de *else if* (“senão, se”).

```python
if primeira_condição:
    primeiro caminho
elif segunda_condição:
    segundo caminho
else:
    caminho restante
```

````python
numero = 0

if numero > 0:
    print('O número é positivo.')
elif numero < 0:
    print('O número é negativo.')
else:
    print('O número é zero.')
````

### Apenas um caminho da cadeia é executado

Python testa as condições de cima para baixo. Quando encontra a primeira condição verdadeira, executa seu bloco e ignora o restante da cadeia.

````python
nota = 8.5

if nota >= 9:
    print('Conceito A')
elif nota >= 7:
    print('Conceito B')
elif nota >= 5:
    print('Conceito C')
else:
    print('Conceito D')
````

A nota `8.5` também é maior que `5`, mas o bloco `nota >= 7` aparece primeiro e já foi escolhido.

## 6. A ordem das condições importa

Em uma cadeia, coloque primeiro os casos mais específicos. Considere uma média igual a `10`: ela satisfaz tanto `media >= 7` quanto `media == 10`. Para mostrar a mensagem especial, devemos testar `media == 10` primeiro.

````python
media = 10

if media == 10:
    print('Aprovado com distinção.')
elif media >= 7:
    print('Aprovado.')
else:
    print('Reprovado.')
````

Se `media >= 7` viesse primeiro, uma média `10` receberia apenas “Aprovado”, e o caso de distinção nunca seria alcançado.

## 7. Vários `if` ou uma cadeia com `elif`?

Use vários `if` quando as verificações forem independentes e mais de um bloco puder ser executado.

````python
numero = 12

if numero > 0:
    print('É positivo.')

if numero % 2 == 0:
    print('É par.')

if numero > 10:
    print('É maior que 10.')
````

As três condições são verdadeiras, então as três mensagens aparecem. Use `if/elif/else` quando os caminhos forem excludentes e apenas um deles deva ser escolhido.

````python
numero = 12

if numero < 0:
    print('Negativo')
elif numero == 0:
    print('Zero')
else:
    print('Positivo')
````

## 8. Condições compostas

Podemos combinar comparações com `and`, `or` e `not`. No exemplo, a pessoa precisa ter idade dentro do intervalo **e** possuir autorização.

````python
idade = 16
tem_autorizacao = True

if idade >= 14 and idade <= 18 and tem_autorizacao:
    print('Participação liberada.')
else:
    print('Participação não liberada.')
````

Parênteses ajudam a tornar condições maiores mais fáceis de ler.

````python
dia = 'sábado'

if dia == 'sábado' or dia == 'domingo':
    print('É fim de semana.')
else:
    print('É dia útil.')
````

## 9. Recebendo dados com `input()`

A função `input()` mostra uma pergunta e espera que a pessoa digite uma resposta. O valor recebido sempre começa como `str`.

```python
nome = input('Digite seu nome: ')
```

Como células com `input()` interrompem a execução automática do notebook, os exemplos seguintes usam valores já definidos. Comentários mostram como trocar esses valores por entradas interativas.

### Convertendo números

Para fazer comparações numéricas, converta o texto recebido:

- `int(input(...))` para um número inteiro;
- `float(input(...))` para um número decimal.

````python
# Para testar com digitação, substitua a próxima linha por:
# idade = int(input('Digite sua idade: '))
idade = 17

if idade >= 18:
    print('Você tem 18 anos ou mais.')
else:
    print('Você tem menos de 18 anos.')
````

### Preparando respostas de texto

Os métodos `.strip()` e `.upper()` ajudam a padronizar uma resposta:

- `.strip()` remove espaços extras do início e do fim;
- `.upper()` transforma as letras em maiúsculas.

Assim, respostas como `' m '`, `'m'` e `'M'` podem ser tratadas da mesma forma.

````python
resposta_original = ' m '
resposta_padronizada = resposta_original.strip().upper()

print(f'Resposta original: {resposta_original!r}')
print(f'Resposta padronizada: {resposta_padronizada!r}')
````

O `!r` da f-string deixa as aspas visíveis, facilitando a observação dos espaços que existiam no texto original.

## 10. Exemplo completo: comparando dois números

O programa testa três possibilidades: o primeiro é maior, o segundo é maior ou os dois são iguais.

````python
# Para usar entrada interativa, substitua as duas próximas linhas por:
# x = int(input('Digite o primeiro número inteiro: '))
# y = int(input('Digite o segundo número inteiro: '))
x = 14
y = 9

if x > y:
    print('O primeiro número é maior.')
elif y > x:
    print('O segundo número é maior.')
else:
    print('Os números são iguais.')
````

## 11. Exemplo completo: classificação de média

Vamos calcular a média de duas notas e escolher uma mensagem. O caso mais específico, média igual a 10, aparece primeiro.

````python
nota1 = 8.0
nota2 = 7.0
media = (nota1 + nota2) / 2

print(f'Média: {media}')

if media == 10:
    print('Aprovado com distinção.')
elif media >= 7:
    print('Aprovado.')
else:
    print('Reprovado.')
````

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `if nota >= 7` | faltaram os dois-pontos | `if nota >= 7:` |
| instrução sem recuo após `if` | faltou indentação | use quatro espaços antes da instrução |
| `if nota = 7:` | `=` faz atribuição | use `if nota == 7:` |
| `else nota < 7:` | `else` não recebe condição | use `elif nota < 7:` ou apenas `else:` |
| testar `media >= 7` antes de `media == 10` | o caso geral captura a nota 10 | teste primeiro o caso específico |
| comparar `input()` diretamente com número | `input()` retorna texto | use `int()` ou `float()` |
| usar vários `if` para opções excludentes | mais de um bloco pode executar | use uma cadeia `if/elif/else` |

Quando surgir um erro, confira a linha do `if`, os dois-pontos e o alinhamento do bloco seguinte.
