# Aula 08.2: Funções `lambda`

Uma função `lambda` representa uma função pequena por meio de uma única expressão. Ela é útil principalmente quando precisamos passar uma operação simples para outra função, como `map()`, `filter()` ou `sorted()`.

> Compare sempre a versão com `def` e a versão com `lambda`. O objetivo é reconhecer que as duas podem representar o mesmo cálculo.

## 1. Revisão: função com `def`

Uma função simples que calcula o dobro pode ser escrita assim:

````python
def calcular_dobro(numero):
    return numero * 2

print(calcular_dobro(5))
````

Essa função possui um parâmetro e retorna uma expressão curta. Ela pode ser representada por uma `lambda`.

````python
calcular_dobro_lambda = lambda numero: numero * 2

print(calcular_dobro_lambda(5))
````

As duas chamadas retornam `10`. A diferença está na forma de definir a função.

## 2. Sintaxe de uma `lambda`

A forma geral é:

```python
lambda parametros: expressão
```

- `lambda` inicia a função;
- os parâmetros aparecem antes dos dois-pontos;
- a expressão aparece depois dos dois-pontos;
- o resultado da expressão é retornado automaticamente.

Não escrevemos `def`, nome ou `return` dentro da expressão.

### Correspondência entre as formas

```python
def calcular_dobro(numero):
    return numero * 2
```

```python
lambda numero: numero * 2
```

O parâmetro `numero` e a expressão `numero * 2` permanecem. `def`, nome, indentação e `return` deixam de aparecer.

## 3. Guardando uma `lambda` em uma variável

Uma função `lambda` não recebe nome na própria definição. Podemos guardar sua referência em uma variável para chamá-la depois.

````python
dobro = lambda x: 2 * x

print(dobro(5))
print(dobro(12))
````

Isso funciona e ajuda no aprendizado. Porém, se uma função será reutilizada com um nome, `def dobro(x): ...` costuma comunicar melhor a intenção. Lambdas são mais comuns como funções curtas usadas no próprio local.

## 4. Lambdas com um parâmetro

A expressão pode realizar diferentes transformações simples.

````python
quadrado = lambda x: x ** 2
metade = lambda x: x / 2
maiusculo = lambda texto: texto.upper()

print(quadrado(6))
print(metade(15))
print(maiusculo('python'))
````

Cada `lambda` avalia uma única expressão e devolve seu resultado.

## 5. Lambdas com vários parâmetros

Parâmetros são separados por vírgulas, assim como em funções com `def`.

````python
somar = lambda a, b: a + b
multiplicar = lambda a, b: a * b
funcao_linear = lambda x, a, b: a * x + b

print(somar(8, 3))
print(multiplicar(8, 3))
print(funcao_linear(2, 3, 5))
````

Os argumentos continuam associados aos parâmetros pela posição, a menos que sejam enviados por nome.

````python
print(funcao_linear(x=4, a=2, b=1))
````

## 6. Expressão condicional

Uma `lambda` pode escolher entre dois resultados usando uma expressão condicional:

```python
valor_se_verdadeiro if condição else valor_se_falso
```

A ordem é diferente do bloco tradicional `if/else`: o primeiro valor aparece antes do `if`.

### Comparando `def` e `lambda`

````python
def transformar_numero(x):
    if x > 10:
        return 2 * x
    else:
        return x

transformar_lambda = lambda x: 2 * x if x > 10 else x

for numero in [1, 10, 100]:
    print(numero, transformar_numero(numero), transformar_lambda(numero))
````

Para `1` e `10`, a condição é falsa e o valor permanece igual. Para `100`, o retorno é `200`. As duas funções produzem os mesmos resultados.

### Classificando paridade

````python
classificar_paridade = lambda numero: 'par' if numero % 2 == 0 else 'ímpar'

print(classificar_paridade(8))
print(classificar_paridade(11))
````

### Comparando dois valores

Com dois resultados, precisamos descrever corretamente o caso de igualdade. Se `x > y` for falso, `y` pode ser maior **ou igual** a `x`.

````python
comparar = lambda x, y: 'Primeiro argumento é maior' if x > y else 'Segundo argumento é maior ou igual'

print(comparar(10, 4))
print(comparar(4, 10))
print(comparar(10, 10))
````

Para distinguir “maior”, “menor” e “igual”, seriam necessárias condições encaixadas. Nesse caso, uma função com `def`, `if`, `elif` e `else` é mais fácil de ler.

````python
def comparar_tres_casos(x, y):
    if x > y:
        return 'Primeiro argumento é maior'
    elif x < y:
        return 'Segundo argumento é maior'
    else:
        return 'Os argumentos são iguais'

print(comparar_tres_casos(10, 10))
````

## 7. Limites de uma `lambda`

Uma `lambda` aceita apenas **uma expressão**. Ela não possui um bloco indentado com várias instruções.

Uma expressão produz um valor, como:

- `x ** 2`;
- `texto.upper()`;
- `'par' if x % 2 == 0 else 'ímpar'`.

Para laços, várias atribuições, tratamento de erros ou muitas condições, use `def`.

## 8. Uso com `map()`

`map()` aplica uma função a cada item. Uma `lambda` permite escrever uma transformação curta diretamente na chamada.

````python
numeros = [1, 2, 3, 4, 5]
quadrados = list(map(lambda numero: numero ** 2, numeros))

print(f'Originais: {numeros}')
print(f'Quadrados: {quadrados}')
````

Dentro de `map()`, `lambda numero: numero ** 2` é passada sem ser chamada manualmente. O próprio `map()` envia cada número a ela.

````python
palavras = ['python', 'dados', 'escola']
maiusculas = list(map(lambda palavra: palavra.upper(), palavras))

print(maiusculas)
````

### Comparação com compreensão de listas

A mesma transformação pode ser escrita com compreensão.

````python
com_map = list(map(lambda numero: numero * 2, numeros))
com_compreensao = [numero * 2 for numero in numeros]

print(f'map: {com_map}')
print(f'compreensão: {com_compreensao}')
````

As duas formas são válidas. Para uma transformação simples, a compreensão frequentemente é mais fácil de ler em Python.

## 9. Uso com `filter()`

`filter()` mantém os itens para os quais a função retorna `True`.

````python
numeros = [1, 2, 3, 4, 5, 6, 7, 8]
pares = list(filter(lambda numero: numero % 2 == 0, numeros))

print(pares)
````

````python
palavras = ['sol', 'escola', 'computador', 'livro']
longas = list(filter(lambda palavra: len(palavra) > 5, palavras))

print(longas)
````

### Comparação com compreensão de listas

````python
com_filter = list(filter(lambda numero: numero > 4, numeros))
com_compreensao = [numero for numero in numeros if numero > 4]

print(f'filter: {com_filter}')
print(f'compreensão: {com_compreensao}')
````

## 10. Ordenação com `sorted()`

`sorted()` cria uma nova lista ordenada. O parâmetro `key` recebe uma função que informa qual valor deve ser usado como critério. Lambdas são muito úteis nesse caso.

````python
palavras = ['uva', 'computador', 'sol', 'escola']
por_tamanho = sorted(palavras, key=lambda palavra: len(palavra))

print(f'Original: {palavras}')
print(f'Por tamanho: {por_tamanho}')
````

Para cada palavra, a `lambda` retorna seu tamanho. `sorted()` usa esses números como critério, mas mantém as palavras na lista final. A lista original não é modificada.

````python
estudantes = [
    {'nome': 'Ana', 'nota': 8.0},
    {'nome': 'Bruno', 'nota': 6.5},
    {'nome': 'Carla', 'nota': 9.2}
]

por_nota = sorted(estudantes, key=lambda estudante: estudante['nota'], reverse=True)

for estudante in por_nota:
    print(f"{estudante['nome']}: {estudante['nota']}")
````

`reverse=True` solicita ordem decrescente. A `lambda` recebe um dicionário de estudante e retorna apenas sua nota como critério.

## 11. Quando usar `lambda` ou `def`?

| Situação | Escolha comum |
|---|---|
| expressão curta usada no próprio local | `lambda` |
| critério simples de `map`, `filter` ou `sorted` | `lambda` |
| função reutilizada em vários lugares | `def` |
| várias condições ou etapas | `def` |
| necessidade de nome explicativo | `def` |
| lógica que precisa de comentários ou testes isolados | `def` |

> Use `lambda` para simplificar uma ideia que já é simples. Não a use para esconder uma lógica complexa em uma linha.

### Erros comuns

| Código com problema | O que aconteceu | Uma correção |
|---|---|---|
| `lambda x return x * 2` | `lambda` não usa `return` | `lambda x: x * 2` |
| `lambda x x * 2` | faltaram os dois-pontos | `lambda x: x * 2` |
| `lambda (x): x * 2` | parâmetros não precisam de parênteses | `lambda x: x * 2` |
| `lambda x: if x > 0 ...` | expressão condicional começa pelo valor | `lambda x: 'positivo' if x > 0 else 'outro'` |
| esquecer o `else` | a expressão condicional exige dois resultados | acrescente o resultado alternativo |
| `map(lambda x: x * 2)` sem sequência | faltou o iterável | `map(lambda x: x * 2, numeros)` |
| colocar muitas condições encaixadas | a leitura fica difícil | use uma função com `def` |

## 12. Prática guiada

Altere os limites e os valores. Depois, escreva as versões equivalentes com compreensão de listas.

````python
pontuacoes = [35, 72, 48, 90, 61]
limite = 60

aprovadas = list(filter(lambda ponto: ponto >= limite, pontuacoes))
com_bonus = list(map(lambda ponto: ponto + 10, aprovadas))
ordenadas = sorted(com_bonus, reverse=True)

print(f'Originais: {pontuacoes}')
print(f'Aprovadas: {aprovadas}')
print(f'Com bônus: {com_bonus}')
print(f'Ordenadas: {ordenadas}')
````
